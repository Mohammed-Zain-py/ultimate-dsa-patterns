# Partition List

## Problem Statement


Rearrange nodes so that:

All nodes smaller than x come first.

Order inside each partition must stay same.


Input

```text
1 → 4 → 3 → 2 → 5 → 2

x = 3
```


Output

```text
1 → 2 → 2 → 4 → 3 → 5
```


Explanation

```text
Split list.

Join partitions.
```

## Problem Explanation


Original

```text
1 → 4 → 3 → 2 → 5 → 2
```


Target


```text
Small

1 → 2 → 2


Large

4 → 3 → 5
```


Combine


```text
1 → 2 → 2 → 4 → 3 → 5
```


Observation


Do NOT sort.


Keep original order.

## Brute Force


Store values.


Separate.


Build again.


Extra memory.

````python
values = [1, 4, 3, 2, 5, 2]

x = 3

left = []

right = []

for value in values:

    if value < x:

        left.append(value)

    else:

        right.append(value)

print(left + right)
````

## Complexity

```text
Time  → O(N)

Space → O(N)
```

## Optimal Approach


Maintain


```text
small

large
```


For each node


```text
Attach

↓

Move
```


Finally


```text
small + large
```

## Dry Run


Initial

```text
1 → 4 → 3 → 2 → 5 → 2
```


Build


```text
Small

1 → 2 → 2


Large

4 → 3 → 5
```


Join


```text
1 → 2 → 2 → 4 → 3 → 5
```

````python
class Node:

    def __init__(self, data=0):

        self.data = data

        self.next = None


class Solution:

    def partition(self, head, x):

        small_dummy = Node()

        large_dummy = Node()

        small = small_dummy

        large = large_dummy

        while head:

            if head.data < x:

                small.next = head

                small = small.next

            else:

                large.next = head

                large = large.next

            head = head.next

        large.next = None

        small.next = large_dummy.next

        return small_dummy.next
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```
