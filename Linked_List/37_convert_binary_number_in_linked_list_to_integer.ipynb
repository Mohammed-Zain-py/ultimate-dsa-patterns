# Convert Binary Number in Linked List to Integer

## Problem Statement

A Linked List represents a binary number.

Each node contains either **0** or **1**.

Return its decimal value.


Input

```text
1 → 0 → 1
```


Output

```text
5
```


Explanation

```text
Binary

101

↓

Decimal

5
```

## Problem Explanation


Binary Number

```text
1 → 0 → 1
```


Equivalent

```text
101₂
```


Calculation

```text
1 × 2²

+

0 × 2¹

+

1 × 2⁰

=

5
```


Efficient Idea

```text
Current Answer

↓

Multiply by 2

↓

Add Current Bit
```

## Brute Force


Store all bits.


Convert to string.


Convert the string

into an integer.


Simple.

Uses extra memory.

````python
bits = [1, 0, 1]

binary = "".join(map(str, bits))

print(int(binary, 2))
````

## Complexity

```text
Time  → O(N)

Space → O(N)
```

## Optimal Approach


Maintain


```text
answer = 0
```


For every node


```text
answer

↓

answer × 2

↓

answer + bit
```


Continue

until the list ends.

## Dry Run


Initial

```text
answer = 0
```


Read

```text
1

↓

0 × 2 + 1

=

1
```


Read

```text
0

↓

1 × 2 + 0

=

2
```


Read

```text
1

↓

2 × 2 + 1

=

5
```

````python
class Solution:

    def getDecimalValue(self, head):

        answer = 0

        while head:

            answer = answer * 2 + head.val

            head = head.next

        return answer
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```

## Pattern Recognition

```text
Traverse

↓

Accumulate

↓

Return
```

## Takeaway

```text
Multiply

↓

Shift Left

↓

Add Bit

↓

Continue
```


This problem demonstrates how a Linked List can represent a binary number and how incremental accumulation eliminates the need to store the entire binary string first.
