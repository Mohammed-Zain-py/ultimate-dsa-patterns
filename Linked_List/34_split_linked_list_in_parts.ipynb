# Split Linked List in Parts

## Problem Statement

Split the Linked List into **k consecutive parts**.

The sizes of the parts should differ by **at most one**.

Earlier parts should contain one extra node if needed.


Input

```text
1 → 2 → 3 → 4 → 5 → 6 → 7

k = 3
```


Output

```text
Part 1

1 → 2 → 3

Part 2

4 → 5

Part 3

6 → 7
```


Explanation

```text
Distribute nodes

as evenly as possible.

The first parts

receive extra nodes.
```

````python
values = [1, 2, 3, 4, 5, 6, 7]

k = 3

size = len(values) // k

extra = len(values) % k

print(size, extra)
````

```text

```

## Dry Run


Original

```text
1 → 2 → 3 → 4 → 5 → 6 → 7
```


Split

```text
Part 1

1 → 2 → 3
```


```text
Part 2

4 → 5
```


```text
Part 3

6 → 7
```

````python
class Solution:

    def splitListToParts(self, head, k):

        length = 0

        current = head

        while current:

            length += 1

            current = current.next

        part_size = length // k

        extra = length % k

        result = []

        current = head

        for i in range(k):

            part_head = current

            size = part_size + (1 if i < extra else 0)

            for _ in range(size - 1):

                if current:

                    current = current.next

            if current:

                nxt = current.next

                current.next = None

                current = nxt

            result.append(part_head)

        return result
````

## Complexity

```text
Time  → O(N)

Space → O(K)
```
