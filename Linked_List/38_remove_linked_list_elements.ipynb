# Remove Linked List Elements

## Problem Statement

Remove every node whose value equals **val**.


Input

```text
1 → 2 → 6 → 3 → 4 → 5 → 6

val = 6
```


Output

```text
1 → 2 → 3 → 4 → 5
```


Explanation

```text
Traverse the list.

Whenever a node

contains the target value,

remove it by reconnecting

the surrounding nodes.
```

## Brute Force

Remove matching values.

Simple.

## Optimal Approach


Create


```text
Dummy Node
```


Traverse


```text
Previous

↓

Current
```


If


```text
Current Value

=

Target
```


Reconnect


```text
Previous

↓

Current.next
```


Otherwise


```text
Move Both
```

## Dry Run


Initial

```text
D → 1 → 2 → 6 → 3 → 4 → 5 → 6
```


Delete First 6

```text
D → 1 → 2 ─────→ 3 → 4 → 5 → 6
```


Delete Last 6

```text
D → 1 → 2 → 3 → 4 → 5
```


Return

```text
1 → 2 → 3 → 4 → 5
```

````python
class Solution:

    def removeElements(self, head, val):

        dummy = ListNode(0)

        dummy.next = head

        previous = dummy

        current = head

        while current:

            if current.val == val:

                previous.next = current.next

            else:

                previous = current

            current = current.next

        return dummy.next
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```

## Takeaway

```text
Dummy

↓

↓

```


Using a dummy node eliminates special handling for deleting the head node, making the algorithm clean and consistent for every deletion.
