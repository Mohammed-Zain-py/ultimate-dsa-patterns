# Reorder Linked List

## Problem Statement


Reorder list into:


```text
L0 → Ln → L1 → Ln-1 → ...
```


Input

```text
1 → 2 → 3 → 4 → 5
```


Output

```text
1 → 5 → 2 → 4 → 3
```


Explanation

```text
Take from front.

Take from back.

Repeat.
```

## Brute Force


Store nodes.


Take:


```text
left

right

left

right
```


Build again.


Extra memory.

````python
values = [1, 2, 3, 4, 5]

result = []

left = 0

right = len(values) - 1

while left <= right:

    result.append(values[left])

    left += 1

    if left <= right:

        result.append(values[right])

        right -= 1

print(result)
````

## Complexity


## Dry Run


Split

```text
1 → 2 → 3

4 → 5
```


Reverse

```text
5 → 4
```


Merge

```text
1 → 5

↓

1 → 5 → 2

↓

1 → 5 → 2 → 4

↓

1 → 5 → 2 → 4 → 3
```

````python
class Solution:

    def reorderList(self, head):

        slow = fast = head

        while fast and fast.next:

            slow = slow.next

            fast = fast.next.next

        prev = None

        current = slow.next

        slow.next = None

        while current:

            nxt = current.next

            current.next = prev

            prev = current

            current = nxt

        first = head

        second = prev

        while second:

            temp1 = first.next

            temp2 = second.next

            first.next = second

            second.next = temp1

            first = temp1

            second = temp2
````
