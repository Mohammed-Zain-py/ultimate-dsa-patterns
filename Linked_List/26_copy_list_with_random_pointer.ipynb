# Copy List with Random Pointer

## Problem Statement

You are given a Linked List where every node has:

- next pointer
- random pointer

Create a deep copy of the entire list.


Input

```text
A → B → C

next pointers

A → B → C → NULL

random pointers

A.random → C
B.random → A
C.random → B
```


Output

```text
A' → B' → C'

A'.random → C'
B'.random → A'
C'.random → B'
```


Explanation

```text
Create completely new nodes.

Do NOT reuse original nodes.

Both next and random pointers
must point to copied nodes.
```

## Dry Run


Original

```text
A → B → C
```


Interleave

```text
A → A' → B → B' → C → C'
```


Assign Random

```text
A'.random → C'

B'.random → A'

C'.random → B'
```


Detach

```text
Original

A → B → C


Copy

A' → B' → C'
```

````python
class Solution:

    def copyRandomList(self, head):

        if not head:
            return None

        current = head

        while current:

            copy = Node(current.val)

            copy.next = current.next

            current.next = copy

            current = copy.next

        current = head

        while current:

            if current.random:

                current.next.random = current.random.next

            current = current.next.next

        current = head

        copy_head = head.next

        while current:

            copy = current.next

            current.next = copy.next

            if copy.next:

                copy.next = copy.next.next

            current = current.next

        return copy_head
````
