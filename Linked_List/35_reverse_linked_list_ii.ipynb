# Reverse Linked List II

## Problem Statement

Reverse the Linked List from position **left** to **right**.

Keep the remaining nodes unchanged.


Input

```text
1 → 2 → 3 → 4 → 5

left = 2

right = 4
```


Output

```text
1 → 4 → 3 → 2 → 5
```


Explanation

```text
Reverse only

the specified part

of the list.
```

## Problem Explanation


Original

```text
1 → 2 → 3 → 4 → 5
```


Reverse Only

```text
2 → 3 → 4
```


Result

```text

```


Observation


Unlike the previous reverse problem,

## Brute Force


Store values.


Reverse only

the selected range.


Rebuild the list.


Uses extra memory.

````python
values = [1, 2, 3, 4, 5]

left = 2

right = 4

values[left - 1:right] = reversed(values[left - 1:right])

print(values)
````

Step 1

## Dry Run


Original

```text
1 → 2 → 3 → 4 → 5
```


Reverse

```text
2 ← 3 ← 4
```


Reconnect Front

```text
1 → 4
```


Reconnect Tail

```text
1 → 4 → 3 → 2 → 5
```

````python
class Solution:

    def reverseBetween(self, head, left, right):

        if not head or left == right:

            return head

        dummy = ListNode(0)

        dummy.next = head

        prev = dummy

        for _ in range(left - 1):

            prev = prev.next

        current = prev.next

        for _ in range(right - left):

            nxt = current.next

            current.next = nxt.next

            nxt.next = prev.next

            prev.next = nxt

        return dummy.next
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```
