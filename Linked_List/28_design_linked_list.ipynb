# Design Linked List

## Problem Statement

Design your own Linked List.

Support the following operations:

- Get
- Insert at Head
- Insert at Tail
- Insert at Index
- Delete at Index


Input

```text
LinkedList()

addAtHead(10)

addAtTail(20)

addAtIndex(1, 15)

get(1)

deleteAtIndex(2)
```


Output

```text
15
```


Explanation

```text
Implement every operation

using Linked List concepts,

not Python lists.
```

## Optimal Approach


Maintain


```text
Head Pointer
```

```text
Get

Insert

Delete
```


Each operation

updates only the required
connections.

## Dry Run


Initially

```text
NULL
```


Insert Head

```text
10
```


Insert Tail

```text
10 → 20
```


Insert Index

```text
10 → 15 → 20
```


Delete

```text
10 → 15
```

````python
class Node:

    def __init__(self, val=0):

        self.val = val

        self.next = None


class MyLinkedList:

    def __init__(self):

        self.head = None


    def get(self, index):

        current = self.head

        while current and index:

            current = current.next

            index -= 1

        return current.val if current else -1


    def addAtHead(self, val):

        node = Node(val)

        node.next = self.head

        self.head = node


    def addAtTail(self, val):

        node = Node(val)

        if not self.head:

            self.head = node

            return

        current = self.head

        while current.next:

            current = current.next

        current.next = node
````

## Complexity

```text
Get             → O(N)

Insert Head     → O(1)

Insert Tail     → O(N)

Delete          → O(N)
```
