# Remove Duplicates from Sorted List II

## Problem Statement

Given a sorted Linked List,

delete **every value that appears more than once**.

Only distinct values should remain.


Input

```text
1 → 2 → 3 → 3 → 4 → 4 → 5
```


Output

```text
1 → 2 → 5
```


Explanation

```text
Remove every node

having duplicate values.

Do not keep

even one copy.
```

## Brute Force


Count frequencies.


Build a new list

using values

whose frequency is 1.


Uses extra memory.

````python
from collections import Counter

values = [1, 2, 3, 3, 4, 4, 5]

freq = Counter(values)

result = [x for x in values if freq[x] == 1]

print(result)
````

## Optimal Approach


Create


```text
Dummy Node
```


Whenever duplicates appear


```text
Skip

Entire Block
```


Otherwise


```text
Move Forward
```

## Dry Run


Initial

```text
1 → 2 → 3 → 3 → 4 → 4 → 5
```


Skip

```text
3 → 3
```


Skip

```text
4 → 4
```


Result

```text
1 → 2 → 5
```

````python
class Solution:

    def deleteDuplicates(self, head):

        dummy = ListNode(0)

        dummy.next = head

        prev = dummy

        current = head

        while current:

            if current.next and current.val == current.next.val:

                while current.next and current.val == current.next.val:

                    current = current.next

                prev.next = current.next

            else:

                prev = prev.next

            current = current.next

        return dummy.next
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```
