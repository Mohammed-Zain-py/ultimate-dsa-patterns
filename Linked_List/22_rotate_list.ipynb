# Rotate Linked List

## Problem Statement

Rotate the list to the right by k positions.


Input

```text
1 → 2 → 3 → 4 → 5

k = 2
```


Output

```text
4 → 5 → 1 → 2 → 3
```


Explanation

```text
Move last k nodes

↓

Bring them to front
```

## Problem Explanation


Original

```text
1 → 2 → 3 → 4 → 5
```


Rotate Once

```text
5 → 1 → 2 → 3 → 4
```


Rotate Twice

```text
4 → 5 → 1 → 2 → 3
```


Observation


Rotating one by one is wasteful.


Idea


```text
Find Length

↓

Make Cycle

↓

Break at right place
```

## Brute Force


Repeat k times:


```text
Remove Tail

↓

Move Front
```


Too many traversals.

````python
values = [1, 2, 3, 4, 5]

k = 2

for _ in range(k):

    values.insert(0, values.pop())

print(values)
````

## Complexity

```text
Time  → O(K × N)

Space → O(N)
```

## Optimal Approach


Step 1

```text
Find Length
```


Step 2

```text
tail.next = head
```


Step 3

```text
Move

length - (k % length)
```


Step 4

```text
Break Cycle
```

## Dry Run


Initial

```text
1 → 2 → 3 → 4 → 5
```


Make Circle

```text
1 → 2 → 3 → 4 → 5
↑               ↓
└───────────────┘
```


Break

```text
4 → 5 → 1 → 2 → 3
```

````python
class Solution:

    def rotateRight(self, head, k):

        if not head or not head.next:
            return head

        length = 1

        tail = head

        while tail.next:

            tail = tail.next

            length += 1

        k %= length

        if k == 0:
            return head

        tail.next = head

        steps = length - k

        new_tail = tail

        while steps:

            new_tail = new_tail.next

            steps -= 1

        new_head = new_tail.next

        new_tail.next = None

        return new_head
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```

## Pattern Recognition

```text
Length

↓

Cycle

↓

Break
```

## Takeaway

```text
Linear

↓

Circular

↓

Linear
```


Rotation becomes easy when the list temporarily becomes circular.
