# Sort Linked List

## Problem Statement

Sort a Linked List in ascending order.


Input

```text
4 → 2 → 1 → 3
```


Output

```text
1 → 2 → 3 → 4
```


Explanation

```text
Sort the nodes.

Do not create
a new list.

Rearrange
the existing nodes.
```

## Dry Run


Original

```text
4 → 2 → 1 → 3
```


Split

```text
4 → 2

1 → 3
```


Sort

```text
2 → 4

1 → 3
```


Merge

```text
1 → 2 → 3 → 4
```

````python
class Solution:

    def sortList(self, head):

        if not head or not head.next:
            return head

        slow = head

        fast = head.next

        while fast and fast.next:

            slow = slow.next

            fast = fast.next.next

        mid = slow.next

        slow.next = None

        left = self.sortList(head)

        right = self.sortList(mid)

        return self.merge(left, right)

    def merge(self, a, b):

        dummy = ListNode()

        tail = dummy

        while a and b:

            if a.val < b.val:

                tail.next = a

                a = a.next

            else:

                tail.next = b

                b = b.next

            tail = tail.next

        tail.next = a or b

        return dummy.next
````
