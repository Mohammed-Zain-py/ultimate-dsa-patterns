# Reverse Nodes in K Group

## Problem Statement


Reverse nodes in groups of size k.

If remaining nodes are less than k:

Leave them unchanged.


Input

```text
1 → 2 → 3 → 4 → 5

k = 2
```


Output

```text
2 → 1 → 4 → 3 → 5
```


Explanation

```text
Reverse blocks.

Not entire list.
```

## Brute Force


Store values.


Reverse each block.


Build again.


Extra memory.

````python
values = [1, 2, 3, 4, 5]

k = 2

result = []

for i in range(0, len(values), k):

    block = values[i:i+k]

    if len(block) == k:

        block.reverse()

    result.extend(block)

print(result)
````

## Optimal Approach


Repeat:


```text
Take K

↓

Reverse

↓

Reconnect
```


Need:


```text
previous_group

current

kth
```

## Dry Run


Initial

```text
1 → 2 → 3 → 4 → 5
```


Reverse First Block

```text
2 → 1 → 3 → 4 → 5
```


Reverse Second Block

```text
2 → 1 → 4 → 3 → 5
```


Remain

```text
5
```


Return.

````python
class Solution:

    def reverseKGroup(self, head, k):

        dummy = Node(0)

        dummy.next = head

        group_prev = dummy

        while True:

            kth = group_prev

            for _ in range(k):

                kth = kth.next

                if not kth:
                    return dummy.next

            group_next = kth.next

            prev = group_next

            current = group_prev.next

            while current != group_next:

                nxt = current.next

                current.next = prev

                prev = current

                current = nxt

            temp = group_prev.next

            group_prev.next = kth

            group_prev = temp
````
