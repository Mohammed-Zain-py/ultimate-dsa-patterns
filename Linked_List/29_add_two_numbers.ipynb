# Add Two Numbers

## Problem Statement

Two non-empty Linked Lists represent two non-negative integers.

Each node stores one digit.

Digits are stored in reverse order.

Return the sum as a Linked List.


Input

```text
List 1

2 → 4 → 3

List 2

5 → 6 → 4
```


Output

```text
7 → 0 → 8
```


Explanation

```text
342

+

465

=

807

Store answer

in reverse order.
```

## Optimal Approach


Maintain


```text
carry
```


At every step


```text
digit1

+

digit2

+

carry
```


Create new node.


Continue until


```text
Both lists

AND

carry

finish.
```

## Dry Run


Initial

```text
2 + 5 = 7
carry = 0
```


Next

```text
4 + 6 = 10

Digit = 0

Carry = 1
```


Next

```text
3 + 4 + 1 = 8
```


Result

```text
7 → 0 → 8
```

````python
class Solution:

    def addTwoNumbers(self, l1, l2):

        dummy = ListNode()

        current = dummy

        carry = 0

        while l1 or l2 or carry:

            total = carry

            if l1:

                total += l1.val

                l1 = l1.next

            if l2:

                total += l2.val

                l2 = l2.next

            carry = total // 10

            current.next = ListNode(total % 10)

            current = current.next

        return dummy.next
````
