# Odd Even Linked List

## Problem Statement

Group all nodes at odd positions together.

Then place all nodes at even positions.

Maintain the original relative order within each group.


Input

```text
1 → 2 → 3 → 4 → 5
```


Output

```text
1 → 3 → 5 → 2 → 4
```


Explanation

```text
Positions are considered.

Not node values.

Odd positions first,

then even positions.
```


```text

```

## Brute Force


Store nodes in

two separate lists.


Join them later.


Uses extra memory.

````python
values = [1, 2, 3, 4, 5]

odd = values[::2]

even = values[1::2]

print(odd + even)
````

## Complexity

```text
Time  → O(N)

Space → O(N)
```

## Optimal Approach


Maintain


```text
Odd Pointer

Even Pointer

Even Head
```


Reconnect


```text
Odd → Next Odd

↓

Even → Next Even
```


Finally


```text
Odd Tail

↓

Even Head
```

## Dry Run


Initial

```text
1 → 2 → 3 → 4 → 5
```


Separate

```text
Odd

1 → 3 → 5


Even

2 → 4
```


Reconnect

```text
1 → 3 → 5 → 2 → 4
```

````python
class Solution:

    def oddEvenList(self, head):

        if not head or not head.next:
            return head

        odd = head

        even = head.next

        even_head = even

        while even and even.next:

            odd.next = even.next

            odd = odd.next

            even.next = odd.next

            even = even.next

        odd.next = even_head

        return head
````

## Complexity

```text
Time  → O(N)

Space → O(1)
```

## Pattern Recognition

```text

↓

Maintain Order

↓

```

## Takeaway

```text
Odd Chain

+

Even Chain

↓

Combine
```


This problem demonstrates how multiple pointer chains can be maintained independently and merged efficiently without using extra memory.
