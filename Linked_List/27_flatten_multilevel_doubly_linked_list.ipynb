# Flatten Multilevel Doubly Linked List

## Problem Statement

A doubly linked list may contain child pointers.

Flatten the list into a single-level doubly linked list.


Input

```text
1 ⇄ 2 ⇄ 3 ⇄ 4
      |
      7 ⇄ 8
          |
          11 ⇄ 12
```


Output

```text
1 ⇄ 2 ⇄ 7 ⇄ 8 ⇄ 11 ⇄ 12 ⇄ 3 ⇄ 4
```


Explanation

```text
Whenever a child exists,

visit the child list first,

then continue with the next node.
```

## Problem Explanation


Normal Doubly Linked List

```text
prev ⇄ node ⇄ next
```


Multilevel List

```text
next

1 ⇄ 2 ⇄ 3
      │
      ▼
      7 ⇄ 8
          │
          ▼
         11 ⇄ 12
```


Goal

```text
Remove every child level

↓

Create one continuous list
```


Traversal Order

```text
Node

↓

Child

↓

Next
```

## Brute Force


Perform DFS.


Store nodes in a list.


Reconnect every node afterwards.


Easy to understand.

Uses extra memory.

stack = ["1", "2", "7", "8", "11", "12", "3", "4"]

print(stack)

## Complexity

```text
Time  → O(N)

Space → O(N)
```

## Optimal Approach


Use DFS.


When a child exists:


```text
Save next

↓

Flatten child

↓

Attach child

↓

Reconnect saved next
```


Flow

```text
Node

↓

Child

↓

Next
```

## Dry Run


Original

```text
1 ⇄ 2 ⇄ 3
      │
      ▼
      7 ⇄ 8
```


Flatten Child

```text
1 ⇄ 2 ⇄ 7 ⇄ 8
```


Reconnect

```text
1 ⇄ 2 ⇄ 7 ⇄ 8 ⇄ 3
```


Continue

```text
Repeat until

child = NULL
```

````python
class Solution:

    def flatten(self, head):

        if not head:
            return head

        stack = [head]

        prev = None

        while stack:

            current = stack.pop()

            if prev:

                prev.next = current

                current.prev = prev

            if current.next:

                stack.append(current.next)

            if current.child:

                stack.append(current.child)

                current.child = None

            prev = current

        return head
````

## Complexity

```text
Time  → O(N)

Space → O(N)
```

## Pattern Recognition

```text

↓

↓

```

## Takeaway

```text
Save Next

↓

Visit Child

↓

Reconnect

↓

Continue
```


Flattening a multilevel linked list is a DFS problem disguised as a Linked List problem. It teaches how to combine pointer manipulation with graph/tree traversal ideas.
