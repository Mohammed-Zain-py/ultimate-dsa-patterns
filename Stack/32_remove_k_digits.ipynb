# Remove K Digits

## Problem Statement

Given a number represented
as a string,

remove exactly `k` digits

so that the resulting number
is as small as possible.

Example:

num = "1432219"

k = 3


Output:

"1219"

# Problem Explanation

We want the smallest
possible number.

If a digit is greater than
the digit immediately after it,

removing the larger digit
can make the number smaller.

Example:

143


The `4` is greater than `3`.

Remove `4`:

13


Therefore:

A decreasing pattern creates
a useful removal opportunity.

# Why Use a Stack?

We process digits
from left → right.

Maintain a Stack of digits
that forms the smallest
possible prefix.

When:

top > current digit

and

we still have removals,

remove the top digit.


This is a
Monotonic Increasing Stack.

# Approach

For every digit:

while:

Stack is not empty

AND

k > 0

AND

stack[-1] > digit


pop the Stack.

Decrease k.


Then push the digit.


After processing all digits,

if k is still greater than 0,

remove digits from the end.

# Dry Run

num = "1432219"

k = 3


1

Stack:

1


4

Stack:

1 4


3

4 > 3

Remove 4.

k = 2

Stack:

1 3


2

3 > 2

Remove 3.

k = 1

Stack:

1 2


2

2 is not greater than 2.

Push.

Stack:

1 2 2


1

2 > 1

Remove 2.

k = 0

Stack:

1 2 1 9


Result:

1219

````python
class Solution:

    def removeKdigits(self, num, k):

        stack = []

        for digit in num:

            while (
                stack
                and k > 0
                and stack[-1] > digit
            ):
                stack.pop()
                k -= 1

            stack.append(digit)

        while k > 0:
            stack.pop()
            k -= 1

        result = "".join(stack).lstrip("0")

        return result if result else "0"


obj = Solution()

print(
    obj.removeKdigits(
        "1432219",
        3
    )
)
````

# Why Is This Greedy?

At every position,

we try to remove a larger
digit before a smaller digit.

Example:

```text
... 8 3 ...
Removing 8
makes the number smaller
at the earliest possible
position.
The Stack lets us repeatedly
make this local decision.
```

# Edge Case

num = "10200"

k = 1


Remove:

1


Result:

0200


Leading zeroes must be removed.

Final:

200


If everything becomes zero,

return:

0

# Complexity

Let:

n = number of digits


Time Complexity:

O(n)


Every digit is pushed once
and popped at most once.


Space Complexity:

O(n)

# Pattern Takeaway

Remove K Digits

→

Greedy

+

Monotonic Increasing Stack


Signal:

A larger previous value
should be removed when a
smaller current value appears.
