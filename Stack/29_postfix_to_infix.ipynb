# Postfix to Infix

## Problem Statement

Convert a Postfix expression
into an Infix expression.

Postfix:

A B +


Infix:

(A + B)

# Problem Explanation

In Postfix,

the operator comes after
its operands.

Example:

A B +


The `+` means:

A + B


For:

A B + C *


First:

A + B


Then:

(A + B) * C

# Approach

Scan the expression:

left → right


If token is an operand:

→ push it.


If token is an operator:

→ pop two expressions.


Create:

(second popped)
operator
(first popped)


Wrap the result in
parentheses.


Push the new expression.

# Why Is Operand Order Important?

Expression:

A B -


When `-` is reached:

first popped = B

second popped = A


Therefore:

A - B


Not:

B - A

# Dry Run

Expression:

A B + C *


A

Stack:

[A]


B

Stack:

[A,B]


+

Pop B

Pop A


Create:

(A+B)


Stack:

[(A+B)]


C

Stack:

[(A+B),C]


*

Create:

((A+B)*C)


Answer:

((A+B)*C)

````python
class Solution:

    def postfixToInfix(self, expression):

        stack = []

        for token in expression.split():

            if token.isalnum():
                stack.append(token)

            else:
                first = stack.pop()
                second = stack.pop()

                result = "(" + second + token + first + ")"

                stack.append(result)

        return stack[-1]


obj = Solution()

print(
    obj.postfixToInfix(
        "A B + C *"
    )
)
````

# Why Use a Stack?

A Postfix operator tells us
when two previous expressions
must be combined.

The Stack stores those
unfinished expressions.

When an operator appears:

two expressions

↓

combine

↓

one new expression

# Another Example

Postfix:

A B C * +


Process:

B C *

→

(B*C)


Then:

A (B*C) +

→

(A+(B*C))


Final:

(A+(B*C))

# Complexity

Let:

n = number of tokens


Time Complexity:

O(n)


Each token is processed once.


Space Complexity:

O(n)


The Stack stores
intermediate expressions.

# Pattern Takeaway
```text
Postfix → Infix

Scan:

left → right


Operand:

→ push


Operator:

→ pop two

→ combine

→ push result


Operand order:

second popped
operator
first popped
```
