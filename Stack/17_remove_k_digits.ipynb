# Remove K Digits

# Problem Statement

Given a string `num` representing a non-negative integer and an integer `k`, remove exactly `k` digits from the number so that the resulting number is the smallest possible.

Return the resulting number as a string.

### Input

Two values:

```text
num
k
```

### Output

The smallest possible number after removing exactly `k` digits.

### Examples

```text
Input:
num = "1432219"
k = 3

Output:
"1219"
```

Explanation:

Remove:

```text
4
3
2
```

The remaining number is:

```text
1219
```

---

```text
Input:
num = "10200"
k = 1

Output:
"200"
```

Removing the first `1` gives:

```text
0200
```

which represents:

```text
200
```

---

```text
Input:
num = "10"
k = 2

Output:
"0"
```

# Problem Explanation

The goal is:

```text
Make the number as small as possible.
```

Suppose we have:

```text
1432219
```

and:

```text
k = 3
```

Look at:

```text
14
```

If we remove `4`, the number becomes:

```text
132219
```

which is smaller than keeping `4`.

The general observation is:

> If a digit is greater than the digit immediately after it, removing that larger digit makes the number smaller.

For example:

```text
43
```

Removing `4` is better than removing `3`.

Therefore, whenever we see:

```text
previous digit > current digit
```

we should consider removing the previous digit.

A Stack is ideal for keeping the digits that currently form the smallest possible prefix.

# Brute Force

A simple approach would be to try different combinations of `k` digits to remove.

For a number with `N` digits, there can be many possible combinations.

The number of combinations is:

```text
C(N, K)
```

which becomes extremely large as `N` increases.

We therefore need a greedy approach.

The Monotonic Stack gives us a way to make the decision locally while guaranteeing the smallest possible result.

# Greedy Observation

Consider:

```text
num = "1432219"
```

Start processing from left to right.

Stack:

```text
1
```

Next digit:

```text
4
```

Since:

```text
4 > 1
```

we keep `4` for now.

Stack:

```text
[1,4]
```

Next digit:

```text
3
```

Now:

```text
4 > 3
```

Removing `4` makes the number smaller.

So:

```text
[1,4]
```

becomes:

```text
[1]
```

and then we add:

```text
3
```

Stack:

```text
[1,3]
```

This process continues.

The Stack is therefore kept in increasing order whenever removals are available.

# Optimal Approach

Maintain a Stack of digits.

For every digit:

```text
while:
    k > 0
    and Stack is not empty
    and Stack top > current digit

    remove Stack top
    decrement k
```

Then push the current digit.

Why?

Because if:

```text
Stack top > current digit
```

the larger digit appears before the smaller digit.

Removing the larger digit makes the number smaller as early as possible.

After processing all digits, there may still be removals left.

This happens when the number is already increasing.

For example:

```text
123456
```

There is no digit where:

```text
previous > current
```

So if we still need to remove digits, remove them from the **end**.

Finally:

```text
remove leading zeros
```

and if nothing remains:

```text
return "0"
```

# Algorithm

```text
Create empty Stack

For every digit:

    While:
        k > 0
        Stack is not empty
        Stack top > current digit

        Pop Stack
        k -= 1

    Push current digit

If k > 0:

    Remove k digits from the end

Remove leading zeros

If result is empty:

    return "0"

Return result
```

````python
class Solution:

    def removeKdigits(self, num: str, k: int) -> str:

        stack = []

        for digit in num:

            while k > 0 and stack and stack[-1] > digit:
                stack.pop()
                k -= 1

            stack.append(digit)

        if k > 0:
            stack = stack[:-k]

        result = ''.join(stack).lstrip('0')

        return result if result else '0'
````

# Why Do We Remove From the Left Side?

This is the most important idea in the problem.

Compare:

```text
327
```

If we can remove one digit:

Remove `3`:

```text
27
```

Remove `2`:

```text
37
```

Remove `7`:

```text
32
```

The smallest result is:

```text
27
```

Why?

Because the leftmost digit has the greatest influence on the numerical value.

Therefore, when we find:

```text
larger digit → smaller digit
```

we should remove the larger digit as early as possible.

That is exactly what the Stack allows us to do.

# When K Remains

Consider:

```text
num = "123456"
k = 2
```

The number is already increasing:

```text
1 < 2 < 3 < 4 < 5 < 6
```

There is no reason to remove an earlier digit because doing so would make the first digit larger.

So after processing:

```text
Stack = [1,2,3,4,5,6]
k = 2
```

We remove from the end:

```text
[1,2,3,4]
```

Result:

```text
"1234"
```

Therefore:

```text
If k remains after the scan,
remove from the right.
```

# Common Mistakes

### Mistake 1 — Removing the Smaller Digit

For:

```text
43
```

removing `3` gives:

```text
4
```

while removing `4` gives:

```text
3
```

We want the smaller result.

Therefore remove the larger preceding digit.

---

### Mistake 2 — Forgetting the Remaining K

After processing the entire number, `k` may still be greater than zero.

In that case:

```python
stack = stack[:-k]
```

removes digits from the end.

---

### Mistake 3 — Returning Leading Zeros

```text
0200
```

should become:

```text
200
```

---

### Mistake 4 — Returning an Empty String

If all digits are removed, the answer is:

```text
"0"
```

not:

```text
""
```

---

### Mistake 5 — Treating This as a Normal Next/Previous Element Problem

There is no explicit:

```text
next greater
```

or:

```text
previous smaller
```

request.

Instead, the Stack is being used to maintain the **smallest possible prefix**.

This is a greedy Monotonic Stack problem.

# Complexity

Let:

```text
N = len(num)
```

Every digit is:

```text
Pushed once
Popped at most once
```

Therefore:

```text
Time → O(N)
```

The Stack can contain at most `N` digits:

```text
Space → O(N)
```

So:

```text
Time  → O(N)
Space → O(N)
```

# Why Is It O(N)?

The `while` loop does not make the algorithm O(N²).

Every digit can be removed from the Stack at most once.

Therefore:

```text
Maximum pushes = N
Maximum pops   = N
```

Total Stack operations:

```text
O(N)
```

This is the same amortized reasoning used throughout Monotonic Stack problems.

# Comparison

| Approach | Time | Space |
|---|---:|---:|
| Brute Force | Exponential / Very High | Depends on approach |
| Greedy + Monotonic Stack | O(N) | O(N) |

The Stack provides an efficient way to make the earliest possible digits as small as possible.

# Takeaway

`Remove K Digits` demonstrates a different use of the Monotonic Stack.

Instead of finding a neighboring element, the Stack maintains the smallest possible number as we scan from left to right.

Core rule:

```text
If stack top > current digit
and k > 0:

    remove stack top
```

Then:

```text
If removals remain:
    remove from the end

Remove leading zeros
Return "0" if necessary
```

Complexity:

```text
Time  → O(N)
Space → O(N)
```

The main interview insight is:

```text
Want smallest sequence
       ↓
Make earlier positions as small as possible
       ↓
Remove larger previous elements
       ↓
Greedy Monotonic Stack
```
