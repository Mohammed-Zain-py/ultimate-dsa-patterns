# Problem Explanation

Python's `list` already provides the operations needed to implement a Stack.

We use the **end of the list as the top**.

```text
Stack

↓

Python List

[10, 20, 30]
          ↑
         TOP
```

Therefore:

```text
append() → Push

pop()    → Pop

stack[-1] → Peek
```

Why use the end?

Because adding and removing from the end of a Python list is efficient.

Avoid using:

```python
insert(0, value)
pop(0)
```

for a Stack.

Those operations require shifting elements and are `O(N)`.

````python
stack = []

stack.append(10)

stack.append(20)

stack.append(30)

print(stack)
````

# Brute Force

There is no meaningful brute-force algorithm for implementing a Stack.

The simplest implementation is to directly use Python's `list` operations.

The important part is choosing the correct end of the list.

```text
End of List

↓

TOP
```

Using the beginning of the list would make Push and Pop slower.

# Complexity

Using the end of a Python list:

| Operation | Time |
|---|---:|
| Push | O(1) amortized |
| Pop | O(1) |
| Peek | O(1) |
| Is Empty | O(1) |
| Size | O(1) |

Space:

```text
O(N)
```

where `N` is the number of elements.

# Implementing a Stack Class

Instead of directly using a list everywhere, we can hide the implementation behind a Stack class.

The user interacts with:

```text
push()
pop()
peek()
is_empty()
size()
```

The internal list remains an implementation detail.

```text
User

↓

Stack Interface

↓

Python List
```

````python
class Stack:

    def __init__(self):
        self.items = []

    def push(self, value):
        self.items.append(value)

    def pop(self):
        if self.is_empty():
            return None

        return self.items.pop()

    def peek(self):
        if self.is_empty():
            return None

        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)
````

````python
stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)

print("Top:", stack.peek())

print("Popped:", stack.pop())

print("Top:", stack.peek())

print("Size:", stack.size())

print("Empty:", stack.is_empty())
````

# Comparison

| Approach | Push | Pop | Peek | Extra Space |
|---|---:|---:|---:|---:|
| List at end | O(1) amortized | O(1) | O(1) | O(N) |
| List at beginning | O(N) | O(N) | O(1) | O(N) |

For a Python Stack:

```text
Use the END of the list as TOP.
```

# Takeaway

A Stack does not require a special built-in data structure.

Python's list can efficiently implement Stack behavior when we use its end as the top.

```text
append()
    ↓
  PUSH

  TOP
   ↓

  POP
   ↑
pop()
```

The important implementation decision is:

```text
TOP → END OF LIST
```

This gives constant-time Stack operations.
