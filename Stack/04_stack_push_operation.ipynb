# Stack Push Operation

# Problem Statement

Implement the `push` operation of a Stack.

The operation must:

1. Add a new element to the top of the Stack.
2. Preserve the LIFO structure.
3. Update the Stack correctly after every insertion.

Input

```text
Stack:

10 → 20 → 30

Push:

40
```

Output

```text
40 → 30 → 20 → 10
```

Explanation

The new element is always inserted at the top.

```text
Before

TOP
 ↓
30 → 20 → 10


Push(40)


After

TOP
 ↓
40 → 30 → 20 → 10
```

# Problem Explanation

The Push operation adds one element to the Stack.

For a Linked List implementation, the new node is inserted at the head.

```text
Before

TOP
 ↓
30 → 20 → 10 → NULL
```

Create:

```text
40
```

Connect it:

```text
40 → 30 → 20 → 10 → NULL
```

Then move the top pointer:

```text
TOP
 ↓
40 → 30 → 20 → 10 → NULL
```

The important order is:

```text
Create Node

↓

Connect Node to Current Top

↓

Move Top to New Node
```

We must connect the new node before changing the top pointer.

# Brute Force

A simple way to demonstrate Push behavior is to use a Python list.

```python
stack = []

stack.append(10)
stack.append(20)
stack.append(30)

print(stack)

stack.append(40)

print(stack)
```

The list represents:

```text
[10, 20, 30, 40]
             ↑
            TOP
```

For a real Stack implementation, however, the operation should be hidden behind a `push()` method.

````python
stack = []

stack.append(10)
stack.append(20)
stack.append(30)

print("Before:", stack)

stack.append(40)

print("After:", stack)
````

# Complexity

For a Python list Stack:

```text
Time  → O(1) amortized
Space → O(1)
```

For a Linked List Stack:

```text
Time  → O(1)
Space → O(1)
```

The new element is inserted directly at the top.

No traversal is required.

# Optimal Approach

For a Linked List implementation:

```text
1. Create a new node.

2. Point the new node to the current top.

3. Make the new node the new top.

4. Increase the size.
```

Visual:

```text
Current

TOP
 ↓
30 → 20 → 10
```

Create:

```text
40
```

Connect:

```text
40 → 30 → 20 → 10
```

Update:

```text
TOP
 ↓
40 → 30 → 20 → 10
```


Push `20`

```text

```

Update Top:

```text

```

Push `30`

```text

```

```text
TOP
```

Push `40`



````python
class Node:

    def __init__(self, value):

        self.value = value

        self.next = None


class Stack:

    def __init__(self):

        self.top = None

        self.size = 0

    def push(self, value):

        node = Node(value)

        node.next = self.top

        self.top = node

        self.size += 1
````

````python
stack = Stack()

stack.push(10)

stack.push(20)

stack.push(30)

stack.push(40)

current = stack.top

while current:

    print(current.value)

    current = current.next
````

# Why Is Push O(1)?

Consider:

```text
TOP
 ↓
30 → 20 → 10 → 5 → 2 → 1 → NULL
```

To push `40`, we do not need to inspect any existing node.

We only perform:

```text
Create 40

↓

40.next = TOP

↓

TOP = 40
```

The number of existing nodes does not matter.

Whether the Stack contains:

```text
10 nodes
```

or:

```text
1,000,000 nodes
```

the same number of pointer operations are performed.

Therefore:

```text
Push → O(1)
```

# Common Mistake

Incorrect order:

```python
self.top = node

node.next = self.top
```

This loses the previous top.

After:

```python
self.top = node
```

the new node is already the top.

Then:

```python
node.next = self.top
```

makes the node point to itself.

Correct order:

```python
node.next = self.top

self.top = node
```

Always save the old connection before replacing the top pointer.

# Comparison

| Implementation | Push |
|---|---:|
| Python List | O(1) amortized |
| Linked List | O(1) |

For the Linked List implementation:

```text
Create Node
     ↓
Connect to Top
     ↓
Update Top
```

No traversal is required.

# Pattern Recognition

When implementing Push using a Linked List:

```text
New Node

↓

New Node.next = Top

↓

Top = New Node
```

This is the same pointer pattern as:

```text
Insert at Beginning
```

from the Linked List module.

The only difference is that the Stack calls this operation:

```text
PUSH
```

# Takeaway

Push means:

```text
Add to TOP
```

For a Linked List Stack:

```text
Create

↓

Connect

↓

Update TOP
```

The operation is:

```text
O(1)
```

The key pointer rule is:

```text
new_node.next = top

```

This simple pointer update is the foundation for every Stack Push operation.
