# Stack Pattern Summary

## Core Idea

A Stack follows:

LIFO

Last In

First Out


Use it when the most recent
item must be processed first.

# Stack Operations

Push:

add to top


Pop:

remove from top


Peek:

view top


Is Empty:

check whether Stack is empty


Size:

number of elements
