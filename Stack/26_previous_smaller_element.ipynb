# Previous Smaller Element

## Problem Statement

You are given an array.

For every element,

find the first element
to its left

that is smaller.

If no smaller element exists,

return:

-1

# Problem Explanation

Example:

arr = [4,5,2,10,8]


For 4:

No element on the left

→ -1


For 5:

Previous smaller = 4


For 2:

No smaller element

→ -1


For 10:

Previous smaller = 2


For 8:

Previous smaller = 2


Output:

[-1,4,-1,2,2]

# Brute Force Approach

For every element,

look at elements
to its left.

Start from the closest
element.

Stop when:

a smaller element

is found.

```python
def previous_smaller_bruteforce(arr):
    n = len(arr)

    result = [-1] * n

    for i in range(n):
        for j in range(i - 1, -1, -1):
            if arr[j] < arr[i]:
                result[i] = arr[j]
                break

    return result


print(
    previous_smaller_bruteforce(
        [4, 5, 2, 10, 8]
    )
)
```

# Complexity of Brute Force

Time Complexity:

O(n²)


Space Complexity:

O(n)


O(n) space is used
for the result array.

# Optimal Idea

The answer must be:

to the left.

So process:

left → right


Maintain a Stack
of possible smaller elements.


For every element:

1. Remove elements
   greater than or equal
   to the current element.

2. The remaining top
   is the previous smaller.

3. Push the current element.

# Why Do We Pop?

Suppose:

current = 10


Stack:

[2,8]


Both are smaller.

The top:

8

is closer to 10.

Therefore:

8 is the previous smaller.


Now suppose:

current = 8

Then:

8 >= 8

so the equal element
cannot be a strictly
smaller answer.

It must be removed.

# Dry Run

arr = [4,5,2,10,8]


4:

Stack = []

Answer = -1

Push 4


5:

4 < 5

Answer = 4

Push 5


2:

5 >= 2 → pop

4 >= 2 → pop

Answer = -1

Push 2


10:

2 < 10

Answer = 2

Push 10


8:

10 >= 8 → pop

2 < 8

Answer = 2


Final:

[-1,4,-1,2,2]

```python
class Solution:
    def previousSmallerElement(self, arr):
        n = len(arr)

        result = [-1] * n

        stack = []

        for i in range(n):

            while stack and stack[-1] >= arr[i]:
                stack.pop()

            if stack:
                result[i] = stack[-1]

            stack.append(arr[i])

        return result


obj = Solution()

print(
    obj.previousSmallerElement(
        [4, 5, 2, 10, 8]
    )
)
```

# Complexity Comparison

Brute Force:

Time → O(n²)

Space → O(n)


Optimal:

Time → O(n)

Space → O(n)

# Important Observation

Previous Smaller Element

→

Left-side search

→

Process left → right

→

Pop elements that are
greater than or equal

→

Stack top becomes
the previous smaller element.

# Pattern Takeaway

Previous Smaller Element

uses a

Monotonic Increasing Stack.

The four basic patterns are now:

Next Greater
Next Smaller
Previous Greater
Previous Smaller
