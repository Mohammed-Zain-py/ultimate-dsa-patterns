# Next Greater Element I

# Problem Statement

The **next greater element** of an element `x` in an array is the first element to the right of `x` that is greater than `x`.

You are given two arrays:

```text
nums1
nums2
```

where `nums1` is a subset of `nums2`.

For every element in `nums1`, find its next greater element in `nums2`.

If no greater element exists, return:

```text
-1
```

### Input

Two integer arrays:

```text
nums1
nums2
```

### Output

Return an array containing the next greater element for every element of `nums1`.

### Example

```text
Input:

nums1 = [4, 1, 2]
nums2 = [1, 3, 4, 2]

Output:

[-1, 3, -1]
```

Explanation:

For `4`:

```text
nums2:
1  3  4  2
      ↑
```

There is no greater element to its right.

```text
4 → -1
```

For `1`:

```text
1  3  4  2
↑  ↑
```

The first greater element is:

```text
3
```

For `2`:

```text
1  3  4  2
         ↑
```

There is no element to its right.

```text
2 → -1
```

# Problem Explanation

The important phrase is:

```text
first greater element to the right
```

Consider:

```text
[2, 1, 5, 3, 4]
```

For `2`:

```text
2 → 5
```

because `5` is the first greater element to its right.

For `1`:

```text
1 → 5
```

For `5`:

```text
5 → -1
```

For `3`:

```text
3 → 4
```

The straightforward solution would compare every element with the elements to its right.

But there is a better pattern.

We can process `nums2` from right to left and maintain a Stack of candidates that could become the next greater element.

This Stack is called a:

```text
Monotonic Stack
```

# Brute Force

For every element, scan all elements to its right until finding a greater value.

Example:

```text
[2, 1, 5, 3, 4]
```

For `2`:

```text
1 → not greater
5 → greater
```

So:

```text
2 → 5
```

For `1`:

```text
5 → greater
```

For `5`:

```text
3 → not greater
4 → not greater
```

No greater element exists:

```text
5 → -1
```

This approach can require scanning almost the entire remaining array for every element.

### Complexity

```text
Time  → O(N²)
Space → O(1)
```

We can improve the time complexity using a Stack.

# Optimal Approach

Process `nums2` from **right to left**.

Maintain a Stack containing possible greater elements.

For each number:

### Remove useless elements

While the Stack is not empty and:

```text
stack.top <= current
```

pop it.

Why?

Because if:

```text
current >= stack.top
```

then that Stack element can never be the next greater element for `current` or for elements further to the left that are smaller than `current`.

It is no longer useful.

### Find the answer

After removing all elements that are smaller than or equal to the current value:

```text
stack.top
```

is the next greater element.

If the Stack is empty:

```text
-1
```

### Add current element

Push the current value onto the Stack.

This keeps the Stack monotonic.

# Why Process From Right to Left?

The problem asks for:

```text
greater element to the RIGHT
```

So when processing an element, everything to its right should already be known.

Example:

```text
2  1  5  3  4
            ↑
```

Start from:

```text
4
```

Then move:

```text
3
```

Then:

```text
5
```

Then:

```text
1
```

Then:

```text
2
```

Therefore, when we reach `1`, the Stack already contains useful information about the elements to its right.

# Monotonic Stack

The Stack maintains values in decreasing order from bottom to top.

For example:

```text
Bottom
  ↓
  5
  2
  ↑
 Top
```

Every time a new value arrives:

```text
while top <= current:
    pop()
```

This removes values that can no longer be useful.

The Stack therefore contains only useful candidates for future next-greater queries.

This is the key idea behind a **Monotonic Stack**.

# Building the Result

While processing `nums2`, create a mapping:

```text
element → next greater element
```

For:

```text
nums2 = [2, 1, 5, 3, 4]
```

we obtain:

```text
{
    2: 5,
    1: 5,
    5: -1,
    3: 4,
    4: -1
}
```

Then process `nums1`.

For:

```text
nums1 = [4, 1, 2]
```

look up:

```text
4 → -1
1 → 5
2 → 5
```

Therefore:

```text
[-1, 5, 5]
```

````python
class Solution:

    def nextGreaterElement(self, nums1: list[int], nums2: list[int]) -> list[int]:

        stack = []
        next_greater = {}

        for num in reversed(nums2):

            while stack and stack[-1] <= num:
                stack.pop()

            if stack:
                next_greater[num] = stack[-1]
            else:
                next_greater[num] = -1

            stack.append(num)

        result = []

        for num in nums1:
            result.append(next_greater[num])

        return result
````

# Edge Cases

### No Greater Element

```text
Input:

nums1 = [4]
nums2 = [1, 2, 3, 4]
```

There is nothing greater than `4` to its right.

```text
Output:

[-1]
```

---

### Immediate Greater Element

```text
[2, 3]
```

For `2`:

```text
2 → 3
```

---

### Greater Element Is Far Away

```text
[2, 1, 1, 1, 5]
```

For `2`:

```text
2 → 5
```

The Stack allows us to skip all the useless smaller values efficiently.

---

### Descending Array

```text
[5, 4, 3, 2, 1]
```

Every element has:

```text
-1
```

as its next greater element.

# Common Mistakes

### Mistake 1 — Processing Left to Right Without the Right Pattern

If we process left to right, the elements to the right have not been seen yet.

For this problem, processing right to left makes the required information available.

---

### Mistake 2 — Not Removing Smaller Elements

Incorrectly keeping:

```text
[5, 2, 1]
```

when processing `3`.

Both:

```text
2
1
```

are useless because:

```text
3
```

is greater than both.

They should be removed.

---

### Mistake 3 — Using `<` Instead of `<=`

The condition should be:

```python
stack[-1] <= num
```

because an equal value is not greater.

For:

```text
[2, 2, 3]
```

the next greater element of the first `2` is:

```text
3
```

not the second `2`.

# Complexity

Let:

```text
N = len(nums2)
M = len(nums1)
```

Each element of `nums2` is:

```text
Pushed once
Popped at most once
```

Therefore the total Stack work is linear.

```text
Time → O(N + M)
```

The Stack and mapping can contain up to `N` elements.

```text
Space → O(N)
```

# Why Is It O(N) and Not O(N²)?

At first glance, this code looks suspicious:

```python
while stack and stack[-1] <= num:
    stack.pop()
```

because there is a loop inside a loop.

But an element cannot be popped repeatedly.

Every element is:

```text
Pushed once
```

and:

```text
Popped at most once
```

Therefore, across the entire algorithm:

```text
Total pushes ≤ N
Total pops   ≤ N
```

So the total work is:

```text
O(N)
```

This is a fundamental property of many Monotonic Stack algorithms.

# Comparison

| Approach | Time | Space |
|---|---:|---:|
| Brute Force | O(N²) | O(1) |
| Monotonic Stack | O(N + M) | O(N) |

The Stack uses additional memory, but reduces the time complexity from quadratic to linear.

# Pattern Recognition

When you see:

```text
Next greater
Next smaller
Previous greater
Previous smaller
```

immediately consider:

```text
Monotonic Stack
```

Typical wording:

```text
first greater element to the right
```

or:

```text
first smaller element to the left
```

is a strong signal.

The direction tells you how to process the array.

```text
Need information on RIGHT
        ↓
Often process RIGHT → LEFT
```

```text
Need information on LEFT
        ↓
Often process LEFT → RIGHT
```

# Takeaway

The most important idea is not the dictionary.

It is the **Monotonic Stack pattern**.

```text
Current element
      ↓
Remove useless candidates
      ↓
Stack top = next useful candidate
      ↓
Record answer
      ↓
Push current element
```

Every element is pushed once and popped at most once.

Therefore:

```text
O(N)
```

Remember this interview trigger:

```text
Next Greater
        ↓
Monotonic Stack
```

This pattern will appear repeatedly in harder Stack problems.
