# Min Stack

# Problem Statement

Design a Stack that supports the following operations:

```text
push(x)
pop()
top()
getMin()
```

All operations must run in:

```text
O(1)
```

### Operations

#### `push(x)`

Push element `x` onto the Stack.

#### `pop()`

Remove the element at the top of the Stack.

#### `top()`

Return the element at the top of the Stack.

#### `getMin()`

Return the smallest element currently present in the Stack.

### Example

```text
push(-2)
push(0)
push(-3)

getMin()
```

Output:

```text
-3
```

Then:

```text
pop()

top()
```

Output:

```text
0
```

Then:

```text
getMin()
```

Output:

```text
-2
```

# Problem Explanation

A normal Stack already gives us:

```text
push() → O(1)
pop()  → O(1)
top()  → O(1)
```

The difficult operation is:

```text
getMin()
```

Suppose the Stack contains:

```text
[5, 2, 8, 1, 4]
```

The minimum is:

```text
1
```

We could scan the entire Stack:

```text
5
2
8
1
4
```

but that takes:

```text
O(N)
```

The problem requires:

```text
getMin() → O(1)
```

Therefore we need to store additional information that allows us to know the minimum immediately.

# Brute Force Approach

Use a normal Stack.

For:

```text
getMin()
```

scan every element and find the smallest value.

Example:

```text
Stack = [5, 2, 8, 1, 4]
```

Scan:

```text
5 → minimum = 5
2 → minimum = 2
8 → minimum = 2
1 → minimum = 1
4 → minimum = 1
```

Result:

```text
1
```

### Complexity

```text
push()   → O(1)
pop()    → O(1)
top()    → O(1)
getMin() → O(N)
```

This does not satisfy the requirement.

We need a way to remember the minimum without scanning the entire Stack.

# Key Insight

We can maintain **two Stacks**.

### Stack 1 — Main Stack

Stores all values.

```text
Main Stack
```

### Stack 2 — Minimum Stack

Stores the minimum value seen at each relevant level.

```text
Min Stack
```

For example, push:

```text
5
```

Main:

```text
[5]
```

Min:

```text
[5]
```

Push:

```text
2
```

Main:

```text
[5,2]
```

Min:

```text
[5,2]
```

Push:

```text
8
```

Main:

```text
[5,2,8]
```

Min:

```text
[5,2]
```

The current minimum is always available at:

```text
min_stack[-1]
```

Therefore:

```text
getMin() → O(1)
```

# Why Does the Minimum Stack Work?

Suppose:

```text
Main Stack:

[5, 2, 8, 1]
```

The minimum is:

```text
1
```

So:

```text
Min Stack:

[5, 2, 1]
```

Now suppose we pop:

```text
1
```

The Main Stack becomes:

```text
[5,2,8]
```

The previous minimum was:

```text
2
```

Because the Min Stack also stores the previous minimum:

```text
[5,2]
```

we immediately recover:

```text
2
```

This is the key idea:

> The Min Stack remembers the minimum that should become active after each important minimum is removed.

# Optimal Approach

Maintain two Stacks:

```python
stack
min_stack
```

### Push

When pushing `x`:

```text
Push x into main Stack.

If min_stack is empty:
    push x

Otherwise:
    push the smaller of:
        x
        current minimum
```

Therefore:

```python
min_stack.append(min(x, min_stack[-1]))
```

### Pop

When removing an element:

```text
pop from stack
pop from min_stack
```

Both Stacks stay synchronized.

### Top

Return:

```python
stack[-1]
```

### Get Minimum

Return:

```python
min_stack[-1]
```

````python
class MinStack:

    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:

        self.stack.append(val)

        if not self.min_stack:
            self.min_stack.append(val)
        else:
            current_min = min(self.min_stack[-1], val)
            self.min_stack.append(current_min)

    def pop(self) -> None:

        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:

        return self.stack[-1]

    def getMin(self) -> int:

        return self.min_stack[-1]
````

# Important Insight

Notice that `min_stack` does not necessarily need to contain only values that are new minimums.

For example:

```text
push(5)
push(2)
push(8)
```

We can store:

```text
Main:
[5,2,8]

Min:
[5,2,2]
```

The top of `min_stack` always tells us:

```text
minimum of everything currently in stack
```

This makes the implementation extremely simple.

An alternative implementation stores only values when they become new minimums, but then the push/pop logic needs additional handling.

For learning purposes, keeping the two Stacks synchronized is easier to reason about.

# Alternative Approach — Store Only Minimum Changes

Instead of storing the minimum at every level, we can store only values that become new minimums.

For:

```text
5, 2, 8, 1
```

Main Stack:

```text
[5,2,8,1]
```

Min Stack:

```text
[5,2,1]
```

When `1` is popped, the previous minimum:

```text
2
```

is revealed.

This uses less auxiliary space when many values are larger than the current minimum.

However, the implementation must carefully determine when to push and pop from the Min Stack.

Both approaches provide:

```text
push   → O(1)
pop    → O(1)
top    → O(1)
getMin → O(1)
```

The synchronized approach is simpler and has predictable `O(N)` auxiliary space.

# Edge Cases

### One Element


Output:

```text

```

---

### Duplicate Minimum

```text
push(2)

push(1)
push(1)
```

The minimum remains:

```text
1
```

After popping one `1`, the minimum must still be:

```text

```

This is why the synchronized Min Stack is useful.

---

### Negative Values

```text
push(-2)

push(-3)
```

Minimum:

```text
-3
```

The algorithm works exactly the same way.

---

### Increasing Values

```text
push(1)
push(2)
push(3)
push(4)
```

Minimum always remains:

```text

```

---

### Decreasing Values

```text
push(4)

push(2)

```

Minimum changes after every push.

# Common Mistakes

### Mistake 1 — Scanning the Main Stack

Doing:

```python
min(stack)
```

inside `getMin()` takes:

```text
O(N)
```

which violates the requirement.

---

### Mistake 2 — Forgetting to Update Min Stack on Pop

When the main Stack is popped:

```text
Min Stack must also be popped.
```

Otherwise the two structures become inconsistent.

---

### Mistake 3 — Mishandling Duplicate Minimums

Consider:

```text
[2, 1, 1]
```

If one `1` is popped, the minimum is still:

```text
1
```

The synchronized Min Stack handles this naturally.

---

### Mistake 4 — Thinking the Minimum Is Always the Bottom

A Stack's bottom element is not necessarily the minimum.

For example:

```text
[5, 8, 2]
```

Minimum:

```text
2
```

Therefore we need a separate structure.

# Complexity

Let:

```text
N = number of elements
```

Each operation performs a constant number of Stack operations.

Therefore:

```text
push()   → O(1)
pop()    → O(1)
top()    → O(1)
getMin() → O(1)
```

The auxiliary Min Stack can contain up to `N` elements.

Therefore:

```text
Space → O(N)
```

Overall:

```text
Time  → O(1) per operation
Space → O(N)
```

# Comparison

| Approach | push | pop | top | getMin | Space |
|---|---:|---:|---:|---:|---:|
| Normal Stack + scan | O(1) | O(1) | O(1) | O(N) | O(N) |
| Stack + Min Stack | O(1) | O(1) | O(1) | O(1) | O(N) |

The second approach satisfies the problem requirement.

# Pattern Recognition

The important signal is:

```text
Stack
+
Need additional information about the current Stack
+
That information must be available in O(1)
```

Think:

```text
Auxiliary Stack
```

For Min Stack:

```text
Main Stack
    +
Minimum Stack
```

The same general technique can be extended to other problems:

```text
Main Stack
    +
Maximum Stack
```

or other state information that can be maintained incrementally.

# Why Not Use a Heap?

A Heap can efficiently maintain the minimum element.

However, a Heap does not naturally follow Stack removal order.

Suppose:

```text
Stack = [5,2,8,1]
```

The top is:

```text
1
```

After popping:

```text
1
```

the next top is:

```text
8
```

But a Min Heap would still want to give us:

```text
2
```

The problem needs both:

```text
Stack behavior
+
Minimum tracking
```

The auxiliary Stack gives us both while preserving LIFO behavior.

# Takeaway

The key idea of Min Stack is:

```text
Normal Stack
     +
Extra Stack storing current minimum
```

At every level:

```text
min_stack[-1]
```

is the minimum of the entire Main Stack.

Therefore:

```text
push   → O(1)
pop    → O(1)
top    → O(1)
getMin → O(1)
```

The major DSA lesson is:

> When a data structure needs an additional query in O(1), consider maintaining auxiliary information alongside the original structure.

For Min Stack:

```text
Main Stack → actual values
Min Stack  → minimum at each level
```

Complexity:

```text
Time  → O(1) per operation
Space → O(N)
```
