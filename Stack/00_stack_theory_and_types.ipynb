# Stack — Theory and Types

# What Is a Stack?

A **Stack** is a linear data structure that follows the:

```text
LIFO

Last In
First Out
```

The element inserted **last** is the first element removed.

A simple real-world example is a stack of plates.

```text
        ┌───────┐
        │ Plate │  ← First removed
        ├───────┤
        │ Plate │
        ├───────┤
        │ Plate │
        ├───────┤
        │ Plate │  ← First inserted
        └───────┘
```

You can only access the top efficiently.

So the Stack mainly works through one end:

```text
        TOP
         ↓
      ┌─────┐
      │  30 │
      ├─────┤
      │  20 │
      ├─────┤
      │  10 │
      └─────┘
```

# Why Does a Stack Exist?

A Stack is useful when the **most recently added item must be processed first**.

Instead of allowing arbitrary access to every element, a Stack restricts access to the top.

```text
Insert

   ↓

  TOP
   ↓
┌─────┐
│  30 │
├─────┤
│  20 │
├─────┤
│  10 │
└─────┘
```

This restriction makes certain problems much easier to model.

Typical situations:

```text
Undo

↓

Backtracking

↓

Function Calls

↓

Expression Evaluation

↓

Parentheses Matching

↓

DFS

↓

Monotonic Stack Problems
```

# LIFO — Last In, First Out

Suppose we insert:

```text
10

20

30
```

The Stack becomes:

```text
        TOP
         ↓
      ┌─────┐
      │  30 │
      ├─────┤
      │  20 │
      ├─────┤
      │  10 │
      └─────┘
```

Now remove one element.

```text
30
```

is removed first.

Then:

```text
20
```

Then:

```text
10
```

Therefore:

```text
Inserted:

10 → 20 → 30

Removed:

30 → 20 → 10
```

That is **LIFO**.

# Core Stack Operations

A Stack mainly provides four operations.

### Push

Add an element to the top.

```text
Before

TOP
 ↓
┌─────┐
│  20 │
├─────┤
│  10 │
└─────┘


Push(30)


After

TOP
 ↓
┌─────┐
│  30 │
├─────┤
│  20 │
├─────┤
│  10 │
└─────┘
```

### Pop

Remove the top element.

```text
30 ← removed
```

### Peek / Top

Look at the top element without removing it.

```text
TOP
 ↓
30
```

### Is Empty

Check whether the Stack contains no elements.

# Stack Operations — Visual Summary

```text
              STACK

                 │
        ┌────────┴────────┐
        ↓                 ↓
      PUSH              POP
        │                 │
        ↓                 ↓
   Add to TOP        Remove TOP


              ┌───────┐
              │ PEEK  │
              └───┬───┘
                  ↓
             View TOP


              ┌──────────┐
              │ IsEmpty  │
              └──────────┘
```

The important rule is:

```text
All normal Stack operations happen at the TOP.
```

# Stack vs Queue

Both are linear data structures.

But they remove elements differently.

### Stack

```text
LIFO

Last In
First Out
```

```text
10 → 20 → 30

Pop

↓

30
```

### Queue

```text
FIFO

First In
First Out
```

```text
10 → 20 → 30

Dequeue

↓

10
```

So:

```text
Stack  → LIFO
Queue  → FIFO
```

This distinction becomes extremely important when choosing a data structure for an interview problem.

# How Can We Implement a Stack?

A Stack is an **abstract data structure**.

That means:

```text
Stack

↓

Defines HOW it behaves

not necessarily HOW it is stored
```

We can implement it using different underlying structures.

Common implementations:

```text
1. Array / Python List

2. Linked List

3. Deque
```

The interface remains:

```text
Push

Pop

Peek

Is Empty
```

Only the underlying implementation changes.

# Stack Using an Array

An array-based Stack stores elements sequentially.

```text
Index

  0     1     2
  ↓     ↓     ↓

┌─────┬─────┬─────┐
│ 10  │ 20  │ 30  │
└─────┴─────┴─────┘
                  ↑
                 TOP
```

In Python, a `list` can be used as a Stack.

```python
stack = []

stack.append(10)
stack.append(20)
stack.append(30)
```

The end of the list represents the top.

```text
append() → Push

pop()    → Pop

stack[-1] → Peek
```

# Stack Using a Linked List

A Stack can also be built using a Linked List.

The head of the Linked List can represent the top.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Push:

```text
New node

↓

Insert at Head
```

Pop:

```text
Remove Head
```

This gives:

```text
Push → O(1)

Pop  → O(1)
```

This is another reason understanding Linked Lists is useful before learning Stack implementations.

# Stack Using Deque

A `deque` supports insertion and deletion from both ends.

For a Stack, we can use one end as the top.

```python
from collections import deque

stack = deque()

stack.append(10)

stack.append(20)

stack.append(30)
```

Then:

```python
stack.pop()
```

removes the top element.

A deque is generally preferred over repeatedly inserting/removing from the beginning of a Python list.

# Complexity

For a properly implemented Stack:

| Operation | Time |
|---|---:|
| Push | O(1) |
| Pop | O(1) |
| Peek | O(1) |
| Is Empty | O(1) |

Space:

```text
O(N)
```

where `N` is the number of elements stored.

# Where Are Stacks Used?

Stacks appear in many important DSA problems.

### 1. Function Calls

Recursive function calls are managed using the **call stack**.

```text
Function A

↓

Function B

↓

Function C
```

The most recent function call must finish first.

```text
C

↓

B

↓

A
```

### 2. Undo / Redo

```text
Action 1

↓

Action 2

↓

Action 3
```

Undo removes:

```text
Action 3
```

first.

### 3. Browser History

Recently visited pages can be tracked using Stack-like behavior.

### 4. Parentheses

```text
( [ { } ] )
```

A Stack helps match opening and closing brackets.

### 5. DFS

Depth First Search can be implemented using a Stack.

### 6. Monotonic Stack

Used for problems such as:

```text
Next Greater Element

Next Smaller Element

Daily Temperatures

Stock Span

Largest Rectangle in Histogram
```

# Stack vs Linked List

A Linked List and Stack are not competing concepts.

They describe different things.

```text
Linked List

↓

How data is connected
```

```text
Stack

↓

How data is accessed
```

A Stack can actually be implemented using a Linked List.

So:

```text
Linked List = underlying structure

Stack = behavior / ADT
```
