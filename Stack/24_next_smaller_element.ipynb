# Next Smaller Element

## Problem Statement

You are given an array.

For every element,

find the first element
to its right

that is smaller.

If no smaller element exists,

return:

-1

# Problem Explanation

Example:

arr = [4,5,2,10,8]


For 4:

Next smaller = 2


For 5:

Next smaller = 2


For 2:

No smaller element

→ -1


For 10:

Next smaller = 8


For 8:

No smaller element

→ -1


Output:

[2,2,-1,8,-1]

# Brute Force Approach

For every element,

look at all elements
to its right.

Stop when:

a smaller element

is found.


Example:

4

↓

5 → not smaller

↓

2 → smaller


Therefore:

4 → 2

```python
def next_smaller_bruteforce(arr):
    n = len(arr)

    result = [-1] * n

    for i in range(n):
        for j in range(i + 1, n):
            if arr[j] < arr[i]:
                result[i] = arr[j]
                break

    return result


print(
    next_smaller_bruteforce(
        [4, 5, 2, 10, 8]
    )
)
```

# Complexity of Brute Force

Time Complexity:

O(n²)


Space Complexity:

O(n)


O(n) space is used
for the result array.

# Optimal Idea

We need the:

next smaller element

to the right.


So process the array

from:

right → left


Maintain a Stack
of elements that can
still become answers.


For every element:

1. Remove elements
   greater than or equal
   to the current element.

2. The remaining top
   is the next smaller element.

3. Push the current element.

# Why Do We Pop?

Suppose:

current = 2


Stack:

[8,10]


Both:

8 > 2

10 > 2


Neither can be
the next smaller
element for 2.


So remove them.


Stack:

[]


Answer:

2 → -1

# Dry Run

arr = [4,5,2,10,8]


Start from the right.


8:

Stack = []

Answer = -1

Push 8


10:

8 < 10

Answer = 8

Push 10


2:

10 >= 2 → pop

8 >= 2 → pop

Stack = []

Answer = -1

Push 2


5:

2 < 5

Answer = 2

Push 5


4:

5 >= 4 → pop

2 < 4

Answer = 2


Final:

[2,2,-1,8,-1]

```python
class Solution:
    def nextSmallerElement(self, arr):
        n = len(arr)

        result = [-1] * n

        stack = []

        for i in range(n - 1, -1, -1):

            while stack and stack[-1] >= arr[i]:
                stack.pop()

            if stack:
                result[i] = stack[-1]

            stack.append(arr[i])

        return result


obj = Solution()

print(
    obj.nextSmallerElement(
        [4, 5, 2, 10, 8]
    )
)
```

# Important Observation

For:

Next Smaller Element


Process:

right → left


And maintain a:

Monotonic Increasing Stack.


The Stack helps us remove
elements that cannot possibly
be the answer anymore.

# Pattern Takeaway

Next Smaller Element

→

Right-side search

→

Process from right to left

→

Pop elements that are
greater than or equal

→

Stack top becomes
the next smaller element.
