# Daily Temperatures

# Problem Statement

Given an array `temperatures` representing the daily temperatures, return an array where:

```text
answer[i]
```

is the number of days you have to wait after day `i` to get a warmer temperature.

If there is no future day with a warmer temperature, set:

```text
answer[i] = 0
```

### Input

An array of integers:

```text
temperatures
```

### Output

An array where each position contains the number of days until a warmer temperature occurs.

### Examples

```text
Input:

temperatures = [73,74,75,71,69,72,76,73]

Output:

[1,1,4,2,1,1,0,0]
```

Explanation:

```text
73 → wait 1 day → 74
74 → wait 1 day → 75
75 → wait 4 days → 76
71 → wait 2 days → 72
69 → wait 1 day  → 72
72 → wait 1 day  → 76
76 → no warmer day → 0
73 → no warmer day → 0
```

# Problem Explanation

This is a variation of the **Next Greater Element** problem.

Previously, we wanted:

```text
What is the next greater element?
```

Here, we want:

```text
How many positions away is the next greater element?
```

Consider:

```text
temperatures = [73, 74, 75, 71, 69, 72, 76, 73]
```

For:

```text
73
```

the next warmer temperature is:

```text
74
```

which is:

```text
1 day away
```

For:

```text
75
```

the next warmer temperature is:

```text
76
```

which is:

```text
4 days away
```

So instead of storing the next greater **value**, we store the next greater **index**.

This is where the Stack becomes particularly useful.

# Input

```python
temperatures = [73,74,75,71,69,72,76,73]
```

# Output

```text
[1,1,4,2,1,1,0,0]
```

# Brute Force

For every day, scan all future days until finding a warmer temperature.

For example:

```text
73, 74, 75, 71, 69, 72, 76, 73
↑
```

Start checking:

```text
74 > 73
```

So:

```text
answer[0] = 1
```

For `75`:

```text
71 → not warmer
69 → not warmer
72 → not warmer
76 → warmer
```

Therefore:

```text
answer[2] = 4
```

This approach may scan almost the entire remaining array for every position.

### Complexity

```text
Time  → O(N²)
Space → O(1)
```

We can improve the time complexity with a Monotonic Stack.

# Optimal Approach

Use a **Monotonic Stack**.

Instead of storing temperatures, store their **indices**.

Process the array from left to right.

For every temperature:

```text
current = temperatures[i]
```

Check the indices stored in the Stack.

While:

```text
current > temperatures[stack[-1]]
```

the current temperature is warmer than the temperature represented by the Stack top.

Therefore we have found the answer for that previous day.

The number of days to wait is:

```text
i - stack[-1]
```

Then remove that index from the Stack.

Finally, push the current index.

The Stack contains indices whose warmer future day has not yet been found.

# Why Store Indices?

We need to return:

```text
Number of days
```

not the warmer temperature itself.

Suppose:

```text
temperatures = [73, 74]
```

Indices:

```text
73 → index 0
74 → index 1
```

When `74` is processed:

```text
current index = 1
previous index = 0
```

Therefore:

```text
1 - 0 = 1
```

So:

```text
answer[0] = 1
```

If we stored only the temperatures, we would know:

```text
73 → 74
```

but we would not directly know how many days apart they are.

Therefore:

```text
Store indices
```

is the correct approach.

# Algorithm

```text
Create result array filled with 0

Create empty Stack

For each index i:

    While Stack is not empty
    and current temperature is warmer
    than the temperature at Stack top:

        previous_index = pop()

        result[previous_index] = i - previous_index

    Push i onto Stack

Return result
```

The Stack stores indices of days that are still waiting for a warmer temperature.

````python
class Solution:

    def dailyTemperatures(self, temperatures: list[int]) -> list[int]:

        result = [0] * len(temperatures)

        stack = []

        for i in range(len(temperatures)):

            while stack and temperatures[i] > temperatures[stack[-1]]:

                previous_index = stack.pop()

                result[previous_index] = i - previous_index

            stack.append(i)

        return result
````

# Edge Cases

### Single Day

```text
Input:

[0]
```

There is no future day.

---

### Completely Increasing

```text
Input:
[70,71,72,73]
```

Every day has a warmer next day.

```text
Output:
[1,1,1,0]
```

---

### Completely Decreasing

```text
Input:
[73,72,71,70]
```

No day has a warmer future temperature.

```text
Output:
[0,0,0,0]
```

---

### Equal Temperatures

```text
Input:
[70,70,71]
```

The second `70` is warmer only when `71` appears.

The first `70` must also wait for `71`.

```text
Output:
[2,1,0]
```

An equal temperature is **not** warmer.

# Common Mistakes

### Mistake 1 — Storing Temperatures Instead of Indices

If the answer requires:

```text
Number of days
```

we need positions.

Therefore store:

```python
stack.append(i)
```

not:

```python
stack.append(temperatures[i])
```

---

### Mistake 2 — Using `>=`

The problem asks for a:

```text
warmer
```

temperature.

Equal is not warmer.

Therefore:

```python
temperatures[i] > temperatures[stack[-1]]
```

is correct.

---

### Mistake 3 — Forgetting the Distance Calculation

When a warmer temperature is found:

```python
previous_index = stack.pop()
```

the answer is:

```python
i - previous_index
```

not the temperature difference.

---

### Mistake 4 — Assuming Remaining Stack Elements Need Processing

At the end, any indices still in the Stack have no warmer future temperature.

Their answers are already correctly initialized to:

```text
0
```

# Complexity

Let:

```text
N = len(temperatures)
```

Every index is:

```text
Pushed once
Popped at most once
```

Therefore:

```text
Time → O(N)
```

The Stack can contain up to `N` indices:

```text
Space → O(N)
```

The result array also requires:

```text
O(N)
```

space.

# Why Is It O(N)?

The nested `while` loop may look like it creates an O(N²) algorithm.

It does not.

Every index can enter the Stack only once:

```text
Push → once
```

and can leave the Stack only once:

```text
Pop → at most once
```

So even though there is a `while` loop, the total number of Stack operations across the entire array is linear.

Therefore:

```text
Total work = O(N)
```

This is one of the most important ideas to understand when analyzing Monotonic Stack problems.

# Comparison

| Approach | Time | Space |
|---|---:|---:|
| Brute Force | O(N²) | O(1) |
| Monotonic Stack | O(N) | O(N) |

The Monotonic Stack trades additional memory for a major improvement in time complexity.

# Pattern Recognition

This problem contains several strong Monotonic Stack signals:

```text
Next warmer
```

which is equivalent to:

```text
Next greater
```

The difference is that we need:

```text
Distance to the next greater element
```

Therefore:

```text
Next Greater
      +
Distance
      ↓
Monotonic Stack + Indices
```

When the problem asks:

```text
How many positions until the next greater/smaller element?
```

store indices rather than just values.

# Takeaway

Daily Temperatures is a direct application of the Monotonic Stack pattern.

The key idea is:

```text
Keep unresolved indices in the Stack.
```

When a warmer temperature arrives:

```text
Current index
      ↓
Resolve previous indices
      ↓
Current index - previous index
```

The most important interview signals are:

```text
Next greater
Next warmer
Next smaller
Previous greater
Previous smaller
```

These should immediately make you consider a Monotonic Stack.

Space → O(N)
