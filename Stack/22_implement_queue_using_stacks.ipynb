# Implement Queue Using Stacks

# Problem Statement

Implement a Queue using only Stack operations.

The Queue should support:

```text
push(x)
pop()
peek()
empty()
```

### Queue

A Queue follows:

```text
FIFO
First In, First Out
```

For example:

```text
push(1)
push(2)
push(3)
```

The Queue becomes:

```text
Front → [1, 2, 3] ← Back
```

Therefore:

```text
pop() → 1
```

### Stack

A Stack follows:

```text
LIFO
Last In, First Out
```

If we push:

```text
1
2
3
```

the next element removed is:

```text
3
```

### Goal

Use Stack behavior to implement Queue behavior.

# Problem Explanation

The main difficulty is that Stack and Queue work in opposite ways.

A Stack gives:

```text
LIFO
```

while a Queue requires:

```text
FIFO
```

Suppose we insert:

```text
1
2
3
```

into a Stack:

```text
TOP
 ↓
3
2
1
```

The next element removed would be:

```text
3
```

But a Queue requires:

```text
1
```

So we need to reverse the order.

### Brute Force Idea

We could maintain the elements directly in Queue order and remove from the front.

However, that does not satisfy the requirement of implementing the Queue using Stack operations.

We therefore need to use the behavior of multiple Stacks to create FIFO behavior.

# Key Insight

Use **two Stacks**:

```text
input_stack
output_stack
```

New elements are pushed into `input_stack`.

When we need to remove an element, we use `output_stack`.

Suppose:

```text
input_stack

TOP
 ↓
3
2
1
```

Move everything into `output_stack`:

```text
output_stack

TOP
 ↓
1
2
3
```

The order has been reversed.

Now:

```python
output_stack.pop()
```

returns:

```text
1
```

which is exactly what the Queue requires.

Therefore:

> Moving elements from one Stack to another reverses their order.

### Important Optimization

We only transfer elements when:

```text
output_stack is empty
```

If `output_stack` already contains elements, we use it directly.

This is what gives us **amortized O(1)** performance.

# Optimal Approach

Maintain two Stacks:

```python
input_stack = []
output_stack = []
```

### `push(x)`

Push `x` into `input_stack`.

### `pop()`

If `output_stack` is empty:

1. Move every element from `input_stack` to `output_stack`.
2. Pop from `output_stack`.

### `peek()`

Use the same transfer logic as `pop()`, but return the top element without removing it.

### `empty()`

The Queue is empty only when both Stacks are empty.

### Algorithm

```text
push(x):

    input_stack.append(x)


pop():

    If output_stack is empty:
        Move all elements from input_stack
        to output_stack

    Return output_stack.pop()


peek():

    If output_stack is empty:
        Move all elements from input_stack
        to output_stack

    Return output_stack[-1]


empty():

    Return True if both stacks are empty
```

````python
class MyQueue:

    def __init__(self):
        self.input_stack = []
        self.output_stack = []

    def push(self, x):
        self.input_stack.append(x)

    def pop(self):
        self.move_elements()
        return self.output_stack.pop()

    def peek(self):
        self.move_elements()
        return self.output_stack[-1]

    def empty(self):
        return not self.input_stack and not self.output_stack

    def move_elements(self):
        if not self.output_stack:
            while self.input_stack:
                self.output_stack.append(self.input_stack.pop())
````

# Code Explanation

### `__init__()`

```python
self.input_stack = []
self.output_stack = []
```

We create two empty Stacks.

`input_stack` receives new elements.

`output_stack` provides elements from the front of the Queue.

---

### `push()`

```python
self.input_stack.append(x)
```

Every new element is added to `input_stack`.

There is no need to move anything during `push()`.

---

### `move_elements()`

```python
if not self.output_stack:
    while self.input_stack:
        self.output_stack.append(self.input_stack.pop())
```

We transfer elements only when `output_stack` is empty.

For example:

```text
input_stack  = [1, 2, 3]
output_stack = []
```

After transfer:

```text
input_stack  = []
output_stack = [3, 2, 1]
```

The top of `output_stack` is now `1`.

---

### `pop()`

```python
self.move_elements()
return self.output_stack.pop()
```

First, make sure the front element is available in `output_stack`.

Then remove it.

---

### `peek()`

```python
self.move_elements()
return self.output_stack[-1]
```

Same preparation as `pop()`, but we only look at the top element.

---

### `empty()`

```python
return not self.input_stack and not self.output_stack
```

Both Stacks must be empty for the Queue to be empty.

# Dry Run

Consider:

```text
push(1)
push(2)
push(3)
pop()
peek()
push(4)
pop()
```

### Initially

```text
input_stack  = []
output_stack = []
```

### After `push(1)`

```text
input_stack = [1]
```

### After `push(2)`

```text
input_stack = [1, 2]
```

### After `push(3)`

```text
input_stack = [1, 2, 3]
```

### `pop()`

`output_stack` is empty, so transfer everything:

```text
input_stack  = []
output_stack = [3, 2, 1]
```

The top is `1`.

Therefore:

```text
pop() → 1
```

Now:

```text
output_stack = [3, 2]
```

### `peek()`

`output_stack` is not empty.

Therefore, no transfer is needed.

```text
peek() → 2
```

### `push(4)`

New elements still go into `input_stack`:

```text
input_stack  = [4]
output_stack = [3, 2]
```

### `pop()`

`output_stack` still contains elements, so we do not transfer `4`.

```text
pop() → 2
```

This preserves FIFO order.

````python
queue = MyQueue()

queue.push(1)
queue.push(2)
queue.push(3)

print(queue.pop())      # 1
print(queue.peek())     # 2

queue.push(4)

print(queue.pop())      # 2
print(queue.pop())      # 3
print(queue.pop())      # 4

print(queue.empty())    # True
````

# Edge Cases

### Empty Queue

```text
input_stack  = []
output_stack = []
```

Therefore:

```text
empty() → True
```

---

### One Element

```text
push(10)
pop() → 10
```

---

### Strict FIFO Order

```text
push(1)
push(2)
push(3)
```

must produce:

```text
1 → 2 → 3
```

when elements are removed.

---

### Push After Pop

Consider:

```text
push(1)
push(2)
pop()
push(3)
```

After removing `1`:

```text
output_stack = [2]
input_stack = []
```

After `push(3)`:

```text
output_stack = [2]
input_stack = [3]
```

The next `pop()` must return:

```text
2
```

not `3`.

This is why we must not transfer elements while `output_stack` still contains elements.

# Common Mistakes

### Mistake 1 — Using Only One Stack

A single Stack naturally gives LIFO behavior.

We need FIFO behavior, so two Stacks are used to reverse the order.

---

### Mistake 2 — Transferring Every Time

Do not transfer all elements on every `pop()`.

Use:

```python
if not self.output_stack:
```

---

### Mistake 3 — Using `>=`-style thinking

There is no comparison involved here.

The important condition is simply whether:

```text
output_stack is empty
```

---

### Mistake 4 — Confusing `pop()` and `peek()`

`pop()`:

```text
returns + removes
```

`peek()`:

```text
returns without removing
```

---

### Mistake 5 — Moving Elements in the Wrong Direction

The transfer should be:

```text
input_stack → output_stack
```

when `output_stack` is empty.

We do not move elements back and forth.

# Complexity

Let:

```text
N = number of elements
```

### `push()`

```text
O(1)
```

We simply append to `input_stack`.

### `pop()`

Worst case:

```text
O(N)
```

This happens when all elements need to be transferred.

However, the **amortized complexity is O(1)**.

Each element is transferred from `input_stack` to `output_stack` at most once.

### `peek()`

```text
Worst case → O(N)
Amortized   → O(1)
```

### `empty()`

```text
O(1)
```

### Space

```text
O(N)
```

The two Stacks together contain all Queue elements.

### Final Complexity

```text
push()  → O(1)
pop()   → O(1) amortized
peek()  → O(1) amortized
empty() → O(1)

Space   → O(N)
```

# Why Is It Amortized O(1)?

The `while` loop may look like `O(N)` every time, but an element cannot be transferred repeatedly.

For every element:

```text
input_stack
     ↓
output_stack
     ↓
removed
```

It is transferred at most once.

Therefore, for `N` elements, the total transfer work is only:

```text
O(N)
```

spread across the operations.

So although one individual `pop()` can take `O(N)`, the average cost over a sequence of operations is:

```text
O(1)
```

This is called **amortized analysis**.

# Pattern Recognition & Takeaway

This is a **Data Structure Design** problem.

The key idea is:

```text
Stack
  ↓
LIFO

Transfer to another Stack
  ↓
Reverse order

Two Stacks
  ↓
FIFO behavior
  ↓
Queue
```

The important condition to remember is:

```python
if not self.output_stack:
```

Only transfer elements when the output Stack is empty.

### Interview Answer

If asked:

> How do you implement a Queue using two Stacks?

Answer:

> "I use one Stack for incoming elements and another Stack for outgoing elements. When the outgoing Stack is empty, I transfer all elements from the incoming Stack to it, which reverses their order and gives FIFO behavior. Since each element is transferred at most once, `pop()` and `peek()` are amortized O(1)."

### Final Complexity

```text
Time  → O(1) amortized per operation
Space → O(N)
```

The main lesson:

> Two Stacks can be combined to simulate a Queue because transferring elements from one Stack to another reverses their order.
