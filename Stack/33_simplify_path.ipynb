# Simplify Path

## Problem Statement

Given an absolute Unix-style
file path,

simplify it.

Rules:

`.` means current directory.

`..` means move to parent.

Multiple `/` are treated
as one separator.

Return the canonical path.

# Problem Explanation

Example:

/home/


Output:

/home


Example:

/home/../user


`..` removes:

home


Result:

/user

# Why Use a Stack?

Directories are processed
from left → right.

When we see:

directory

→ enter it.


When we see:

..

→ go back to the previous
directory.


The most recent directory
must be removed first.

That is LIFO behavior.

Therefore:

Stack.

# Approach

Split the path by `/`.

For every part:

If empty:

ignore.


If `.`:

ignore.


If `..`:

if Stack is not empty,

pop.


Otherwise:

push the directory.


At the end:

join the Stack
with `/`.

# Dry Run

Path:

/a/./b/../../c/


Parts:

a

.

b

..

..

c


Process:

a

Stack:

[a]


.

ignore


b

Stack:

[a,b]


..

pop b

Stack:

[a]


..

pop a

Stack:

[]


c

Stack:

[c]


Result:

/c

```python
class Solution:

    def simplifyPath(self, path):

        stack = []

        for part in path.split("/"):

            if part == "" or part == ".":
                continue

            if part == "..":

                if stack:
                    stack.pop()

            else:
                stack.append(part)

        return "/" + "/".join(stack)


obj = Solution()

print(
    obj.simplifyPath(
        "/a/./b/../../c/"
    )
)
```

# Important Edge Case

Path:

/../


There is no parent directory
above the root.

So:

..

does nothing.


Result:

/

# Another Example

Path:

/home//foo/


Repeated `/`

does not create
a directory.

So:

/home//foo/


becomes:

/home/foo/

# Complexity

Let:

n = length of path


Time Complexity:

O(n)


Every character is processed
through the split parts.


Space Complexity:

O(n)


The Stack may contain
all directory names.

# Pattern Takeaway

Directory navigation

+

Nested / recent-state removal

→

Stack


`.`:

ignore


`..`:

pop


directory:

push
