# Infix to Postfix

## Problem Statement

Convert an Infix expression
into a Postfix expression.

Infix:

A + B


Postfix:

A B +

# Problem Explanation

Infix notation is the normal
mathematical notation.

Example:

A + B * C


Because multiplication has
higher precedence,

the expression means:

A + (B * C)


Postfix removes the need
for parentheses and precedence.

Result:

A B C * +

# Operator Precedence

Higher precedence operators
must be processed first.

```text
* / → higher

+ - → lower

Example:
A + B * C
* is processed first.
Therefore:
A B C * +
```

# Approach

Scan the expression
from left → right.


If token is an operand:

→ add to output.


If token is `(`:

→ push to Stack.


If token is `)`:

→ pop until `(`.


If token is an operator:

→ pop operators with
  greater or equal precedence.

→ push current operator.


At the end:

→ pop remaining operators.

# Dry Run

Expression:

A + B * C


A

Output:

A


+

Stack:

+


B

Output:

A B


*

`*` has higher precedence
than `+`.

Push `*`.

Stack:

+ *


C

Output:

A B C


End.

Pop:

*

then:

+


Result:

A B C * +

````python
class Solution:

    def infixToPostfix(self, expression):

        precedence = {
            "+": 1,
            "-": 1,
            "*": 2,
            "/": 2
        }

        output = []

        stack = []

        for token in expression:

            if token.isalnum():
                output.append(token)

            elif token == "(":
                stack.append(token)

            elif token == ")":

                while stack and stack[-1] != "(":
                    output.append(stack.pop())

                stack.pop()

            else:

                while (
                    stack
                    and stack[-1] != "("
                    and precedence[stack[-1]] >= precedence[token]
                ):
                    output.append(stack.pop())

                stack.append(token)

        while stack:
            output.append(stack.pop())

        return " ".join(output)


obj = Solution()

print(
    obj.infixToPostfix(
        "A+B*C"
    )
)
````

# Why Use a Stack?

Operators cannot always
be added immediately.

Example:

A + B * C


When `+` appears,

we do not yet know whether
a higher-priority operator
will appear later.


The Stack temporarily stores
operators until their correct
position in Postfix is known.

# Parentheses

Example:

(A + B) * C


Process:

A

+

B


When `)` appears,

pop until `(`.


Output:

A B +


Then process `*`.

Final:

A B + C *

# Complexity

Let:

n = number of tokens


Time Complexity:

O(n)


Each token is pushed
and popped at most once.


Space Complexity:

O(n)


Output and Stack
can require O(n).

# Pattern Takeaway

Infix → Postfix

uses:

Stack
+
Operator Precedence
+
Parentheses Handling


Operands:

→ Output


Operators:

→ Stack


This is the key pattern
behind expression conversion.
