# Problem Statement

Implement a Stack using only Queue operations.

The Stack should support:

```text
push(x)
pop()
top()
empty()
```

### Stack

A Stack follows:

```text
LIFO
Last In, First Out
```

For example:

```text
push(1)
push(2)
push(3)
```

The Stack becomes:

```text
TOP
 ↓
3
2
1
```

Therefore:

```text
pop() → 3
```

### Queue

A Queue follows:

```text
FIFO
First In, First Out
```

For example:

```text
[1, 2, 3]
 ↑
Front
```

The first element removed is:

```text
1
```

### Goal

Use Queue behavior to implement Stack behavior.

# Problem Explanation

The difficulty is that Stack and Queue have opposite behaviors.

A Stack requires:

```text
LIFO
```

while a Queue naturally provides:

```text
FIFO
```

Suppose we perform:

```text
push(1)
push(2)
push(3)
```

A Stack should remove:

```text
3
```

But a Queue would remove:

```text
1
```

Therefore, we need to rearrange the elements so that the newest element becomes available at the front of the Queue.

The key idea is:

> After adding a new element, move the older elements behind it.

For example:

```text
Before push(3):

[1, 2]
 ↑
Front
```

Add `3`:

```text
[1, 2, 3]
```

Now rotate the older elements:

```text
[2, 3, 1]
```

Then:

```text
[3, 1, 2]
```

Now `3` is at the front.

Therefore:

```text
pop() → 3
```

This gives us Stack behavior using a Queue.

# Brute Force

One approach is to use **two Queues**.

We can maintain:

```text
queue1
queue2
```

When pushing a new element:

1. Add the new element to `queue2`.
2. Move all elements from `queue1` into `queue2`.
3. Swap the two Queues.

For example:

```text
queue1 = [1, 2]
```

Push `3`:

```text
queue2 = [3]
```

Move the old elements:

```text
queue2 = [3, 1, 2]
```

Then swap the Queues.

Now the front is always the most recently inserted element.

Therefore:

```text
pop() → front of queue
```

This works, but it requires two Queues.

We can simplify the implementation by using only **one Queue**.

# Key Insight

We can implement the Stack using **one Queue**.

Suppose the Queue currently contains:

```text
[1, 2]
```

Now we perform:

```text
push(3)
```

First add `3`:

```text
[1, 2, 3]
```

But the front is still `1`.

We need `3` at the front.

So we rotate the first two elements:

```text
[2, 3, 1]
```

Rotate again:

```text
[3, 1, 2]
```

Now:

```text
Front → 3
```

Therefore:

```text
pop() → 3
```

### General Rule

After adding a new element:

```text
Add the element
      ↓
Rotate all previous elements
      ↓
New element moves to the front
```

Then `pop()` becomes very simple:

```python
queue.popleft()
```

The front of the Queue always represents the top of our Stack.

# Optimal Approach

Use one Queue.

### `push(x)`

1. Add `x` to the back of the Queue.
2. Rotate all elements that were already present before `x`.
3. This places `x` at the front.

### `pop()`

Remove and return the front element.

Because the newest element is always at the front:

```text
Queue front = Stack top
```

### `top()`

Return the front element without removing it.

### `empty()`

Check whether the Queue is empty.

### Algorithm

```text
push(x):

    Add x to queue

    Repeat for all elements
    that were already in the queue:

        Remove front element
        Add it to the back


pop():

    Remove and return front element


top():

    Return front element


empty():

    Return whether queue is empty
```

````python
from collections import deque


class MyStack:

    def __init__(self):
        self.queue = deque()

    def push(self, x):
        self.queue.append(x)

        for _ in range(len(self.queue) - 1):
            self.queue.append(self.queue.popleft())

    def pop(self):
        return self.queue.popleft()

    def top(self):
        return self.queue[0]

    def empty(self):
        return not self.queue
````

# Code Explanation

### `__init__()`

```python
self.queue = deque()
```

We use Python's `deque` because it provides efficient operations from both ends.

---

### `push()`

```python
self.queue.append(x)
```

First, add the new element to the back.

For example:

```text
[1, 2]
```

After:

```text
push(3)
```

we have:

```text
[1, 2, 3]
```

But `3` needs to be at the front.

So we rotate the previous elements:

```python
for _ in range(len(self.queue) - 1):
    self.queue.append(self.queue.popleft())
```

The first rotation:

```text
[1, 2, 3]
```

becomes:

```text
[2, 3, 1]
```

The second rotation:

```text
[2, 3, 1]
```

becomes:

```text
[3, 1, 2]
```

Now the newest element is at the front.

---

### `pop()`

```python
return self.queue.popleft()
```

The front of the Queue is always the top of our simulated Stack.

Therefore, removing the front gives the correct Stack behavior.

---

### `top()`

```python
return self.queue[0]
```

The front element represents the Stack top.

We return it without removing it.

---

### `empty()`

```python
return not self.queue
```

If the Queue contains no elements, the Stack is empty.

# Dry Run

Consider:

```text
push(1)
push(2)
push(3)
pop()
top()
```

### Initially

```text
queue = []
```

### `push(1)`

Add `1`:

```text
[1]
```

There are no older elements to rotate.

```text
[1]
```

---

### `push(2)`

Add `2`:

```text
[1, 2]
```

Rotate the previous element:

```text
[2, 1]
```

Now `2` is at the front.

---

### `push(3)`

Add `3`:

```text
[2, 1, 3]
```

Rotate twice:

```text
[1, 3, 2]
```

then:

```text
[3, 2, 1]
```

Now:

```text
Front → 3
```

Therefore:

```text
pop() → 3
```

Queue becomes:

```text
[2, 1]
```

Then:

```text
top() → 2
```

This is exactly Stack behavior:

```text
3 → 2 → 1
```

````python
stack = MyStack()

stack.push(1)
stack.push(2)
stack.push(3)

print(stack.pop())      # 3
print(stack.top())      # 2

stack.push(4)

print(stack.pop())      # 4
print(stack.pop())      # 2
print(stack.pop())      # 1

print(stack.empty())    # True
````

# Edge Cases and Common Mistakes

### Empty Stack

```text
queue = []
```

Therefore:

```text
empty() → True
```

---

### One Element

```text
push(10)
pop() → 10
```

---

### Multiple Pushes

```text
push(1)
push(2)
push(3)
```

The Queue internally becomes:

```text
[3, 2, 1]
```

So:

```text
pop() → 3
pop() → 2
pop() → 1
```

---

### Common Mistake 1 — Forgetting the Rotation

Simply doing:

```python
self.queue.append(x)
```

is not enough.

The newest element would remain at the back, giving FIFO behavior instead of LIFO behavior.

---

### Common Mistake 2 — Rotating the Wrong Number of Times

After adding `x`, rotate exactly:

```text
len(queue) - 1
```

times.

The newly added element itself should not be moved.

---

### Common Mistake 3 — Using `pop()` Before Checking the Problem's Interface

The Queue is being used internally.

The outside interface must still behave like a Stack:

```text
push()
pop()
top()
empty()
```

# Complexity

Let:

```text
N = number of elements
```

### `push()`

After adding an element, we rotate all previous elements.

Therefore:

```text
Time → O(N)
```

### `pop()`

We remove the front element:

```text
Time → O(1)
```

### `top()`

We access the front:

```text
Time → O(1)
```

### `empty()`

Checking whether the Queue is empty:

```text
Time → O(1)
```

### Space

The Queue stores all `N` elements:

```text
Space → O(N)
```

### Final Complexity

```text
push()  → O(N)
pop()   → O(1)
top()   → O(1)
empty() → O(1)

Space   → O(N)
```

The trade-off is intentional:

> We make `push()` expensive so that `pop()` and `top()` become O(1).

# Pattern Recognition

This is another **Data Structure Design** problem.

The previous problem was:

```text
Two Stacks
    ↓
Queue
```

This problem reverses the idea:

```text
One Queue
    ↓
Rearrange elements
    ↓
Stack
```

The important observation is:

```text
Stack needs newest element at the front
```

So after every `push()`:

```text
new element
     ↓
move to front
```

Then the Queue's front behaves exactly like the Stack's top.

### Important Pattern

When implementing one data structure using another, ask:

```text
Which operation should be expensive?

Which operations should be fast?

Can I rearrange the internal structure
during insertion so removal becomes simple?
```

Here we choose:

```text
push() → O(N)
pop()  → O(1)
top()  → O(1)
```

# Takeaway

A Queue normally gives:

```text
FIFO
```

but a Stack requires:

```text
LIFO
```

We solve this by keeping the newest element at the **front** of the Queue.

The process is:

```text
push(x)
   ↓
Add x to back
   ↓
Rotate previous elements
   ↓
x moves to front
   ↓
Front = Stack Top
```

Therefore:

```text
push()  → O(N)
pop()   → O(1)
top()   → O(1)
empty() → O(1)

Space   → O(N)
```
