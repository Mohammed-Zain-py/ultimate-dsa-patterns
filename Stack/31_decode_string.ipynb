# Decode String

## Problem Statement

Given an encoded string,

decode it.

The encoding rule is:

k[encoded_string]

means:

repeat `encoded_string`
exactly `k` times.

Example:

3[a]

Output:

aaa

# Problem Explanation

Example:

3[a2[c]]


First:

2[c]

becomes:

cc


Then:

a + cc

becomes:

acc


Finally:

3 × acc


Result:

accaccacc

# Why Use a Stack?

The string can contain
nested brackets.

Example:

3[a2[c]]


We must remember:

the previous number

and

the previous string


before entering a new
bracket.

A Stack stores these
unfinished states.

# Approach

Scan left → right.


If digit:

build the repeat count.


If `[`:

push:

current number

current string


Then reset both.


If character:

add it to current string.


If `]`:

pop the previous state.

Repeat the current string
and attach it to the previous
string.

# Dry Run

3[a2[c]]


Read 3:

number = 3


Read `[`:

push:

(3, "")


Read a:

current = "a"


Read 2:

number = 2


Read `[`:

push:

(2, "a")


Read c:

current = "c"


Read `]`:

"c" × 2

= "cc"


Previous string:

"a"

So:

"a" + "cc"

= "acc"


Read final `]`:

"acc" × 3

= "accaccacc"

````python
class Solution:

    def decodeString(self, s):

        stack = []

        current = ""

        number = 0

        for char in s:

            if char.isdigit():

                number = number * 10 + int(char)

            elif char == "[":

                stack.append(
                    (current, number)
                )

                current = ""

                number = 0

            elif char == "]":

                previous, repeat = stack.pop()

                current = (
                    previous
                    + current * repeat
                )

            else:

                current += char

        return current


obj = Solution()

print(
    obj.decodeString(
        "3[a2[c]]"
    )
)
````

# Why Store Two Values?

When `[` appears,

we need to remember:

```text
previous string

and
repeat count

Example:

3[ab]
Before entering:

current = ""
number = 3

So Stack stores:

("", 3)
After decoding ab,
we return to the saved state.
```

# Nested Structures

Example:

2[x3[y]]


The Stack handles:

outer `2`

↓

inner `3`

↓

finish inner

↓

return to outer

This is the same LIFO behavior
seen in nested parentheses.

# Complexity

Let:

n = length of the input
and decoded output be m.


Time Complexity:

O(m)


because the decoded result
must be constructed.


Space Complexity:

O(m)


for the output and
Stack states.

# Pattern Takeaway

Nested structure

↓

remember previous state

↓

Stack


Common signal:

```text
[
]
nested
repeat
return to previous state
