# Stack Using Linked List

# Problem Statement

Implement a Stack using a Linked List.

The Stack must support:

```text
Push
Pop
Peek
Is Empty
Size
```

The Stack must follow:

```text
LIFO

Last In
First Out
```

Example

```text
Push 10
Push 20
Push 30
```

Stack

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Pop

```text
30
```

Remaining Stack

```text
TOP
 ↓
20 → 10 → NULL
```

# Problem Explanation

A Stack can be implemented using a Singly Linked List.

The key decision is:

```text
HEAD = TOP
```

So the Stack looks like:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Push happens at the head:

```text
New Node
   ↓
TOP → 30 → 20 → 10
```

Pop also happens at the head:

```text
TOP → 30 → 20 → 10

Remove 30

↓

TOP → 20 → 10
```

Because insertion and deletion at the head of a Linked List are both `O(1)`, Stack operations can also be performed in constant time.

# Why Use the Head as the Top?

We could use the tail as the top.

But that would make Pop expensive in a Singly Linked List.

If the tail is the top:

```text
10 → 20 → 30 → NULL
              ↑
             TOP
```

To remove `30`, we first need to find `20`.

That requires traversal:

```text
10 → 20 → 30
```

Therefore:

```text
Pop → O(N)
```

Instead, use the head:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Now Pop simply removes the first node:

```text
TOP → 20 → 10
```

Therefore:

```text
Push → O(1)
Pop  → O(1)
Peek → O(1)
```

````python
class Node:

    def __init__(self, value):

        self.value = value

        self.next = None
````

# Push

Push adds a new node to the top.

Before:

```text
TOP
 ↓
20 → 10 → NULL
```

Push `30`:

```text
30
 ↓
20 → 10
```

After:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

The new node becomes the new head.

````python
top = Node(20)

top.next = Node(10)

new_node = Node(30)

new_node.next = top

top = new_node

current = top

while current:

    print(current.value)

    current = current.next
````

# Pop

Pop removes the node at the top.

Before:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Remove `30`:

```text
TOP
 ↓
20 → 10 → NULL
```

The top simply moves to:

```text
top.next
```

````python
top = Node(30)

top.next = Node(20)

top.next.next = Node(10)

removed = top

top = top.next

print("Popped:", removed.value)

print("New Top:", top.value)
````

# Peek

Peek returns the value at the top without removing it.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Peek:

```text
30
```

The Stack remains unchanged.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

````python
top = Node(30)

top.next = Node(20)

top.next.next = Node(10)

print("Top:", top.value)
````

# Is Empty

The Stack is empty when:

```text
top == None
```

Empty:

```text
TOP
 ↓
NULL
```

Not Empty:

```text
TOP
 ↓
30 → 20 → NULL
```

````python
top = None

print("Empty:", top is None)

top = Node(10)

print("Empty:", top is None)
````

# Size

Unlike a Python list, a Linked List does not automatically provide its length.

We have two choices:

```text
1. Traverse every time

2. Maintain a size variable
```

Maintaining a size variable is better.

Whenever we Push:

```text
size += 1
```

Whenever we Pop:

```text
size -= 1
```

Therefore:

```text
Size → O(1)
```

# Stack Class

The complete Stack maintains:

```text
top

size
```

Operations:

```text
push()

pop()

peek()

is_empty()

get_size()
```

````python
class Node:

    def __init__(self, value):

        self.value = value

        self.next = None


class Stack:

    def __init__(self):

        self.top = None

        self.size = 0

    def push(self, value):

        node = Node(value)

        node.next = self.top

        self.top = node

        self.size += 1

    def pop(self):

        if self.is_empty():

            return None

        value = self.top.value

        self.top = self.top.next

        self.size -= 1

        return value

    def peek(self):

        if self.is_empty():

            return None

        return self.top.value

    def is_empty(self):

        return self.top is None

    def get_size(self):

        return self.size
````

````python
stack = Stack()

stack.push(10)

stack.push(20)

stack.push(30)

print("Top:", stack.peek())

print("Popped:", stack.pop())

print("Top:", stack.peek())

print("Size:", stack.get_size())

print("Empty:", stack.is_empty())
````

# Complexity

| Operation | Time | Extra Space |
|---|---:|---:|
| Push | O(1) | O(1) |
| Pop | O(1) | O(1) |
| Peek | O(1) | O(1) |
| Is Empty | O(1) | O(1) |
| Size | O(1) | O(1) |

Overall Stack storage:

```text
O(N)
```

where `N` is the number of elements.

# Comparison

| Implementation | Push | Pop | Peek |
|---|---:|---:|---:|
| Python List | O(1) amortized | O(1) | O(1) |
| Linked List | O(1) | O(1) | O(1) |

The main difference is how the underlying data is stored.

```text
List

↓

Contiguous Dynamic Array
```

```text
Linked List

↓

Nodes + Pointers
```

The Stack behavior remains the same:

```text
LIFO
```

# Pattern Recognition

When implementing a Stack with a Singly Linked List:

```text
Need LIFO

↓

Choose Head as TOP

↓

Push at Head

↓

Pop at Head
```

The important rule is:

```text
TOP = HEAD
```

This keeps both Push and Pop at:

```text
O(1)
```

# Takeaway

A Stack can be built directly on top of a Singly Linked List.

The most important design decision is:

```text
HEAD = TOP
```

Then:

```text
Push

↓

Insert at Head
```

```text
Pop

↓

Delete Head
```

Both operations are:

```text
O(1)
```
