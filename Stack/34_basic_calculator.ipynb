# Basic Calculator

## Problem Statement

Given a string containing
a mathematical expression,

evaluate it.

The expression can contain:

- integers
- `+`
- `-`
- `(`
- `)`
- spaces

Example:

1 + 1


Output:

2

# Problem Explanation

Consider:

1 + (2 - 3)


Inside the parentheses:

2 - 3

= -1


Therefore:

1 + (-1)

= 0


The challenge is remembering
the result and sign from the
outer expression while entering
and leaving parentheses.

# Why Use a Stack?

When `(` appears,

the current calculation
must be saved.

When `)` appears,

we return to the previous
calculation.

So we store:

previous result

and

previous sign


The most recently opened
parentheses are resolved first.

Therefore:

Stack.

# Approach

Maintain:

result

number

sign


When digit:

build number.


When `+`:

add:

sign × number

Then set:

sign = +1


When `-`:

add:

sign × number

Then set:

sign = -1


When `(`:

push current result
and current sign.

Reset them.


When `)`:

finish current number.

Pop previous sign
and previous result.

Combine them.

# Dry Run

Expression:

1 + (2 - 3)


Start:

result = 0

sign = +1


Read 1:

result = 1


Read +:

sign = +1


Read (:

push:

(1, +1)


Read 2:

result = 2


Read -:

add 2

sign = -1


Read 3:

result = -1


Read ):

pop:

previous result = 1

previous sign = +1


Combine:

1 + (-1)

= 0

```python
class Solution:

    def calculate(self, s):

        stack = []

        result = 0

        number = 0

        sign = 1

        for char in s:

            if char.isdigit():

                number = (
                    number * 10
                    + int(char)
                )

            elif char in "+-":

                result += sign * number

                number = 0

                if char == "+":
                    sign = 1
                else:
                    sign = -1

            elif char == "(":

                stack.append(result)

                stack.append(sign)

                result = 0

                sign = 1

            elif char == ")":

                result += sign * number

                number = 0

                previous_sign = stack.pop()

                previous_result = stack.pop()

                result = (
                    previous_result
                    + previous_sign * result
                )

        result += sign * number

        return result


obj = Solution()

print(
    obj.calculate(
        "1 + (2 - 3)"
    )
)
```

# Why Store the Sign?

Consider:

1 - (2 + 3)


The whole parenthesized
expression must be subtracted.

So when entering:

(2 + 3)


we remember:

sign = -1


After calculating:

2 + 3 = 5


we return to the outer
expression:

1 - 5

= -4

# Complexity

Let:

n = length of expression


Time Complexity:

O(n)


Each character is processed
once.


Space Complexity:

O(n)


The Stack stores the
nested calculations.

# Important Observation

The Stack is not storing
every number.

It stores the state of the
calculation before entering
parentheses.

That state is:

previous result

+

previous sign

# Pattern Takeaway

Basic Calculator

→

Expression parsing

+

Stack

+

State restoration


When nested expressions
temporarily interrupt the
current calculation,

save the previous state
on a Stack.
