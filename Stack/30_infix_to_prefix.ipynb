# Infix to Prefix

## Problem Statement

Convert an Infix expression
into a Prefix expression.

Infix:

A + B


Prefix:

+ A B

# Problem Explanation

Infix:

A + B * C


Multiplication has higher
precedence.

So:

A + (B*C)


Prefix representation:

+ A * B C

# Approach

A common method is:

1. Reverse the Infix expression.

2. Swap:
   `(` with `)`
   `)` with `(`

3. Convert the reversed
   expression to Postfix.

4. Reverse the resulting
   Postfix expression.

The result is Prefix.

# Dry Run

Infix:

A+B*C


Reverse:

C*B+A


Convert to Postfix:

C B * A +


Reverse:

+ A * B C


Therefore:

Prefix:

+ A * B C

# Parentheses

Example:

(A+B)*C


Reverse:

C*)B+A(


Swap parentheses:

C*(B+A)


Convert to Postfix:

C B A + *


Reverse:

* + A B C


Final Prefix:

* + A B C

```python
class Solution:

    def infixToPrefix(self, expression):

        precedence = {
            "+": 1,
            "-": 1,
            "*": 2,
            "/": 2
        }

        expression = expression[::-1]

        expression = expression.replace(
            "(", "#"
        ).replace(
            ")", "("
        ).replace(
            "#", ")"
        )

        output = []

        stack = []

        for token in expression:

            if token.isalnum():
                output.append(token)

            elif token == "(":
                stack.append(token)

            elif token == ")":

                while stack and stack[-1] != "(":
                    output.append(stack.pop())

                stack.pop()

            else:

                while (
                    stack
                    and stack[-1] != "("
                    and precedence[stack[-1]] > precedence[token]
                ):
                    output.append(stack.pop())

                stack.append(token)

        while stack:
            output.append(stack.pop())

        return " ".join(output[::-1])


obj = Solution()

print(
    obj.infixToPrefix(
        "A+B*C"
    )
)
```

# Why Reverse?

Prefix places the operator
before its operands.

By reversing the expression,

we can temporarily transform
the problem into a Postfix-style
processing problem.

Then reversing the final result
produces Prefix.

# Complexity

Let:

n = number of tokens


Time Complexity:

O(n)


Space Complexity:

O(n)


The expression and Stack
require linear space.

# Prefix vs Infix vs Postfix

Infix:

A + B


Prefix:

+ A B


Postfix:

A B +


Same operation,

different placement
of the operator.

# Pattern Takeaway

Infix → Prefix

Reverse

↓

Swap parentheses

↓

Convert to Postfix

↓

Reverse


Stack handles:

operators

parentheses

precedence
