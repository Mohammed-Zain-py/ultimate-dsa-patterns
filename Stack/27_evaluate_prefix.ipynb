# Evaluate Prefix Expression

## Problem Statement

Given a valid Prefix expression,

evaluate its result.

In Prefix notation,

the operator appears
before its operands.

Example:

+ 2 3

Result:

5

# Problem Explanation

Prefix:

+ 2 3


The operator comes first.

For:

- 2 3

we calculate:

2 - 3

= -1


For:

* 4 5

we calculate:

4 × 5

= 20

# Approach

Process the expression:

right → left


If the token is:

operand

→ push it.


If the token is:

operator

→ pop two operands.


Apply:

first popped
operator
second popped


Then push the result.

# Why Right to Left?

Example:

- + 7 3 2


Read from the right:

2

3

7

+

When `+` is reached,

the required operands
are already on the Stack.

So:

7 + 3

= 10


Then:

10 - 2

= 8

# Dry Run

Expression:

- + 7 3 2


Process right → left:

2

Stack = [2]


3

Stack = [2,3]


7

Stack = [2,3,7]


+

Pop:

7

3


7 + 3 = 10

Stack = [2,10]


-

Pop:

10

2


10 - 2 = 8


Answer:

8

```python
class Solution:
    def evaluatePrefix(self, expression):
        stack = []

        tokens = expression.split()

        for token in reversed(tokens):

            if token.lstrip("-").isdigit():
                stack.append(int(token))

            else:
                first = stack.pop()
                second = stack.pop()

                if token == "+":
                    result = first + second

                elif token == "-":
                    result = first - second

                elif token == "*":
                    result = first * second

                elif token == "/":
                    result = int(first / second)

                stack.append(result)

        return stack[-1]


obj = Solution()

print(
    obj.evaluatePrefix(
        "- + 7 3 2"
    )
)
```

# Important Detail

For Prefix,

we scan:

right → left.


The order of operands matters.

For:

- 10 3


we need:

10 - 3


not:

3 - 10


Therefore:

first = first popped

second = second popped


result:

first operator second

# Complexity

Let:

n = number of tokens


Time Complexity:

O(n)


Each token is processed once.


Space Complexity:

O(n)


The Stack may contain
multiple operands.

# Prefix vs Postfix

Prefix:

operator first

+ 2 3


Postfix:

operator last

2 3 +


Prefix evaluation:

right → left


Postfix evaluation:

left → right


Both use a Stack.

# Pattern Takeaway

Expression Evaluation

→

Stack


Prefix:

right → left


Postfix:

left → right


When an operator appears,

pop the required operands,

calculate,

then push the result.
