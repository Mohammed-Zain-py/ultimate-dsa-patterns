# Stack Size Operation

# Problem Statement

Implement the `size()` operation of a Stack.

The operation must return the number of elements currently stored in the Stack.

Input

```text
Stack

TOP
 ↓
30 → 20 → 10 → NULL
```

Output

```text
3
```

After removing one element:

```text
TOP
 ↓
20 → 10 → NULL
```

The size becomes:

```text
2
```

For an empty Stack:

```text
TOP
 ↓
NULL
```

the size is:

```text
0
```

# Problem Explanation

A Stack implemented using a Linked List does not automatically know how many nodes it contains.

There are two ways to find the size.

### Approach 1 — Traverse the Stack

Start from the top and count every node.

```text
TOP
 ↓
30 → 20 → 10 → NULL

count = 0

30 → count = 1

20 → count = 2

10 → count = 3
```

This requires visiting every node.

```text
Time → O(N)
```

### Approach 2 — Maintain a Size Variable

Keep a variable:

```text
size
```

Whenever we push:

```text
size += 1
```

Whenever we pop:

```text
size -= 1
```

Then:

```text
size()
```

can return the value directly.

```text
Time → O(1)
```

This is the better implementation.

# Brute Force

The direct approach is to traverse the entire Linked List every time `size()` is called.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Start:

```text
count = 0
```

Visit `30`:

```text
count = 1
```

Visit `20`:

```text
count = 2
```

Visit `10`:

```text
count = 3
```

Return:

```text
3
```

The problem is that every call requires traversing the Stack.

```text
Time → O(N)
Space → O(1)
```

````python
class Node:

    def __init__(self, value):

        self.value = value
        self.next = None


def get_size(top):

    count = 0

    current = top

    while current:

        count += 1

        current = current.next

    return count
````

# Optimal Approach

Maintain the Stack size as part of the Stack's state.

Initially:

```text
size = 0
```

When pushing:

```text
size += 1
```

When popping:

```text
size -= 1
```

Therefore, the current size is always available.

```text
Push
 ↓
size + 1

Pop
 ↓
size - 1
```

The `size()` operation simply returns:

```python
self.size
```

````python
class Node:

    def __init__(self, value):

        self.value = value
        self.next = None


class Stack:

    def __init__(self):

        self.top = None
        self.size = 0

    def push(self, value):

        node = Node(value)

        node.next = self.top

        self.top = node

        self.size += 1

    def pop(self):

        if self.is_empty():

            return None

        value = self.top.value

        self.top = self.top.next

        self.size -= 1

        return value

    def size_of_stack(self):

        return self.size

    def is_empty(self):

        return self.top is None
````

````python
stack = Stack()

print("Size:", stack.size_of_stack())

stack.push(10)
stack.push(20)
stack.push(30)

print("Size:", stack.size_of_stack())

stack.pop()

print("Size:", stack.size_of_stack())

stack.pop()

stack.pop()

print("Size:", stack.size_of_stack())
````

# Why Is the Optimal Approach O(1)?

Consider a Stack containing:

```text
1,000,000 elements
```

A traversal-based approach must inspect all of them.

```text
O(N)
```

But with a maintained size variable:

```python
return self.size
```

we directly return the stored value.

The number of elements does not matter.

Therefore:

```text
size() → O(1)
```

# Important Invariant

The `size` variable must always match the actual number of nodes.

This is an invariant:

```text
size == number of nodes in the Stack
```

After every operation:

```text
Push

↓

size increases by 1
```

```text
Pop

↓

size decreases by 1
```

If either update is forgotten, the Stack's reported size becomes incorrect.

# Edge Cases

### Empty Stack

```text
TOP
 ↓
NULL

size = 0
```

### One Element

```text
TOP
 ↓
10 → NULL

size = 1
```

### Multiple Elements

```text
TOP
 ↓
30 → 20 → 10 → NULL

size = 3
```

### Pop Until Empty

```text
30 → 20 → 10

↓

20 → 10

↓

10

↓

NULL
```

The size must change:

```text
3 → 2 → 1 → 0
```

# Common Mistake

Forgetting to update `size` during `push()` or `pop()`.

Incorrect:

```python
def push(self, value):

    node = Node(value)

    node.next = self.top

    self.top = node
```

The node is added, but:

```text
size
```

was not increased.

Correct:

```python
self.size += 1
```

Likewise, Pop must contain:

```python
self.size -= 1
```

# Complexity

| Approach | Time | Space |
|---|---:|---:|
| Traverse every time | O(N) | O(1) |
| Maintain size variable | O(1) | O(1) |

The maintained-size approach uses one additional integer:

```text
O(1)
```

This small amount of extra state eliminates repeated traversal.

# Comparison

| Operation | Effect on Size |
|---|---:|
| Push | `size += 1` |
| Pop | `size -= 1` |
| Peek | No change |
| Is Empty | No change |
| Size | Returns current value |

The important distinction is:

```text
Push / Pop

↓

Modify size
```

```text
Peek / Is Empty / Size

↓

Only inspect state
```

# Pattern Recognition

When implementing a data structure operation that repeatedly asks for:

```text
Current number of elements
```

consider whether maintaining the value directly is better than recalculating it.

For a Linked List Stack:

```text
Maintain:

size
```

Then:

```text
size()

↓

return size
```

This is a common data-structure design technique:

```text
Store useful state

↓

Keep it updated

↓

Answer queries efficiently
```

# Takeaway

A Linked List does not inherently store its length.

For a Stack, maintaining a `size` variable gives:

```text
Push → O(1)
Pop  → O(1)
Size → O(1)
```

The critical invariant is:

```text
size == actual number of elements
```

Whenever the Stack changes, the size must change with it.
