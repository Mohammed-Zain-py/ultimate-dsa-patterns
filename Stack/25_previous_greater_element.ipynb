# Previous Greater Element

## Problem Statement

You are given an array.

For every element,

find the first element
to its left

that is greater.

If no greater element exists,

return:

-1

# Problem Explanation

Example:

arr = [4,5,2,10,8]


For 4:

No element on the left

→ -1


For 5:

Previous greater = -1


For 2:

Previous greater = 5


For 10:

No greater element on the left

→ -1


For 8:

Previous greater = 10


Output:

[-1,-1,5,-1,10]

# Brute Force Approach

For every element,

look at elements
to its left.

Start from the closest
element on the left.

Stop when:

a greater element

is found.

```python
def previous_greater_bruteforce(arr):
    n = len(arr)

    result = [-1] * n

    for i in range(n):
        for j in range(i - 1, -1, -1):
            if arr[j] > arr[i]:
                result[i] = arr[j]
                break

    return result


print(
    previous_greater_bruteforce(
        [4, 5, 2, 10, 8]
    )
)
```

# Complexity of Brute Force

Time Complexity:

O(n²)


Space Complexity:

O(n)


O(n) space is used
for the result array.

# Optimal Idea

The answer must be:

to the left.

So process the array:

left → right


Maintain a Stack
of elements that can
still become answers.


For every element:

1. Remove elements
   smaller than or equal
   to the current element.

2. The remaining top
   is the previous greater
   element.

3. Push the current element.

# Why Do We Pop?

Suppose:

current = 8


Stack:

[10, 12]


Both are greater than 8.

But the top:

12

is closer to 8.

Therefore:

12 is the answer.


For a future element,

if 10 is blocked by 12,
it may still be useful.

The Stack therefore maintains
candidates in decreasing order.

Elements that are smaller than
or equal to the current element
cannot become a previous greater
element for the current element
or future elements above them.

# Dry Run

arr = [4,5,2,10,8]


4:

Stack = []

Answer = -1

Push 4


5:

4 <= 5 → pop

Stack = []

Answer = -1

Push 5


2:

5 > 2

Answer = 5

Push 2


10:

2 <= 10 → pop

5 <= 10 → pop

Answer = -1

Push 10


8:

10 > 8

Answer = 10


Final:

[-1,-1,5,-1,10]

```python
class Solution:
    def previousGreaterElement(self, arr):
        n = len(arr)

        result = [-1] * n

        stack = []

        for i in range(n):

            while stack and stack[-1] <= arr[i]:
                stack.pop()

            if stack:
                result[i] = stack[-1]

            stack.append(arr[i])

        return result


obj = Solution()

print(
    obj.previousGreaterElement(
        [4, 5, 2, 10, 8]
    )
)
```

# Complexity Comparison

Brute Force:

Time → O(n²)

Space → O(n)


Optimal:

Time → O(n)

Space → O(n)


Every element is:

pushed once

and

popped at most once.

# Important Observation

Previous Greater Element

→

Left-side search

→

Process left → right

→

Pop elements that are
smaller than or equal

→

Stack top becomes
the previous greater element.

# Pattern Takeaway

Previous Greater Element

uses a

Monotonic Decreasing Stack.

The direction of processing
depends on where the answer exists.

Previous → left → process left to right.
