# Validate Stack Sequences

# Problem Statement

Given two integer arrays `pushed` and `popped`, determine whether there exists a sequence of stack push and pop operations that produces the `popped` sequence from the `pushed` sequence.

### Input

Two arrays:

```text
pushed
popped
```

`pushed` represents the order in which elements are pushed onto the Stack.

`popped` represents the required order in which elements are removed.

### Output

Return:

```text
True
```

if the `popped` sequence is a valid Stack sequence.

Otherwise return:

```text
False
```

### Examples

```text
Input:
pushed = [1,2,3,4,5]
popped = [4,5,3,2,1]

Output:
True
```

One valid sequence is:

```text
push 1
push 2
push 3
push 4
pop 4
push 5
pop 5
pop 3
pop 2
pop 1
```

---

```text
Input:
pushed = [1,2,3,4,5]
popped = [4,3,5,1,2]

Output:
False
```

# Problem Explanation

A Stack follows:

```text
LIFO
Last In, First Out
```

Therefore, once an element is pushed, it cannot be removed while another element above it remains in the Stack.

For example:

```text
Stack:

1
2
3
4  ← top
```

The only element that can be popped is:

```text
4
```

So when processing the desired `popped` sequence, we need to ask:

```text
Can the current desired value be removed from the Stack?
```

If the Stack top matches the desired value:

```text
pop it
```

Otherwise, we need to continue pushing elements from `pushed`.

If we have pushed everything and the desired value still cannot be found on top:

```text
The sequence is impossible.
```

# Brute Force Idea

One way to think about the problem is to generate possible push/pop combinations and check whether one of them produces `popped`.

However, the number of possible operation sequences can become very large.

We do not need to generate every possibility.

There is a deterministic greedy strategy:

```text
Push elements until the desired popped element
reaches the top.

Then pop it immediately.
```

This works because delaying a valid pop does not help us produce a different valid Stack ordering.

Therefore we can simulate the Stack directly.

# Greedy Observation

Suppose:

```text
pushed = [1,2,3,4,5]
popped = [4,5,3,2,1]
```

We need:

```text
4
```

first.

Push:

```text
1
2
3
4
```

Stack:

```text
[1,2,3,4]
```

Now `4` is on top.

Pop:

```text
[1,2,3]
```

Next desired value:

```text
5
```

Push `5`:

```text
[1,2,3,5]
```

Pop `5`.

Then:

```text
3
2
1
```

can all be popped in order.

The key is:

```text
Push only when necessary.
Pop immediately when the desired value reaches the top.
```

# Optimal Approach

Maintain:

```text
Stack
```

and a pointer:

```text
j
```

to the current element in `popped`.

For every value in `pushed`:

```text
Push it.

Then:

while Stack is not empty
and Stack top == popped[j]:

    Pop Stack
    move j forward
```

At the end:

```text
if j == len(popped):
    True
else:
    False
```

If every desired pop was successfully matched, the sequence is valid.

# Algorithm

```text
Create empty Stack
j = 0

For value in pushed:

    Push value

    While:
        Stack is not empty
        and j < len(popped)
        and Stack top == popped[j]:

        Pop Stack
        j += 1

Return:

    j == len(popped)
```

````python
class Solution:

    def validateStackSequences(
        self,
        pushed: list[int],
        popped: list[int]
    ) -> bool:

        stack = []
        j = 0

        for value in pushed:

            stack.append(value)

            while (
                stack
                and j < len(popped)
                and stack[-1] == popped[j]
            ):
                stack.pop()
                j += 1

        return j == len(popped)
````

# Why the Greedy Approach Works

The important question is:

```text
When the desired popped value is on top,
should we pop it immediately?
```

Yes.

Suppose:

```text
Stack top == popped[j]
```

If we do not pop it, any future push will place another element above it.

That cannot make the current desired element easier to pop.

Therefore the safest and simplest decision is:

```text
If the desired element is on top,
pop it immediately.
```

This gives us a deterministic simulation instead of trying every possible sequence of operations.

# Common Mistakes

### Mistake 1 — Checking Only the Final Stack

The important thing is not just whether the final Stack is empty.

We need to verify whether the required **pop order** can be produced.

---

### Mistake 2 — Popping the Wrong Element

A Stack can only pop:

```text
stack[-1]
```

We cannot search the Stack and remove an arbitrary element.

---

### Mistake 3 — Continuing to Push After Everything Is Popped

Once:

```text
j == len(popped)
```

all required elements have already been removed.

The result is already valid.

---

### Mistake 4 — Forgetting the Pointer

The Stack tells us what is available.

The pointer `j` tells us:

```text
Which element must be popped next?
```

Both pieces of state are necessary.

# Complexity

Let:

```text
N = len(pushed)
```

Every element is:

```text
Pushed once
Popped at most once
```

Therefore:

```text
Time → O(N)
```

The Stack can contain at most `N` elements:

```text
Space → O(N)
```

So:

```text
Time  → O(N)
Space → O(N)
```

# Why Is It O(N)?

The `while` loop is nested inside the `for` loop, but it does not make the algorithm O(N²).

Every value can be popped only once.

Therefore:

```text
Maximum pushes = N
Maximum pops   = N
```

Total operations are linear:

```text
O(N)
```

Again, this is amortized analysis.

# Comparison

| Approach | Time | Space |
|---|---:|---:|
| Generate possible operation sequences | Exponential / Very High | High |
| Stack Simulation | O(N) | O(N) |

The Stack simulation is deterministic and directly models the actual LIFO constraint.

# Pattern Recognition

This problem is different from:

```text
Next Greater
Previous Smaller
Monotonic Stack
```

Instead, the signal is:

```text
Given push order
+
Given pop order
+
Determine whether the pop order is possible
```

Think:

```text
Simulate the Stack
```

The general pattern is:

```text
Process push sequence
        ↓
Push each element
        ↓
Check desired pop
        ↓
If top matches → pop
        ↓
Continue
```

Whenever a problem asks whether a sequence can be produced by a Stack, think about **direct Stack simulation**.
