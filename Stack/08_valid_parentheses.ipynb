# Valid Parentheses

## Problem Statement

Given a string `s` containing only the characters:

```text
'('  ')'  '{'  '}'  '['  ']'
```

determine if the input string is valid.

A string is valid if:

1. Every opening bracket has a corresponding closing bracket of the same type.

2. Opening brackets must be closed in the correct order.

3. Every closing bracket must have a corresponding opening bracket.

### Input

A string `s`.

### Output

Return:

```text
True
```

if the string is valid.

Otherwise return:

```text
False
```

### Examples

```text
Input:
s = "()"

Output:
True
```

```text
Input:
s = "()[]{}"

Output:
True
```

```text
Input:
s = "(]"

Output:
False
```

```text
Input:
s = "([)]"

Output:
False
```

```text
Input:
s = "{[]}"

Output:
True
```

## Problem Explanation

The important part of this problem is the phrase:

```text
closed in the correct order
```

Consider:

```text
([ ])
```

The last opening bracket is:

```text
[
```

Therefore it must be the first bracket closed.

```text
(
[
]
)
```

This is exactly:

```text
LIFO
```

The most recently opened bracket must be closed first.

That makes a Stack the natural data structure for this problem.

### Bracket Pairs

```text
( → )
[ → ]
{ → }
```

When we see an opening bracket:

```text
(
[
{
```

we push it onto the Stack.

When we see a closing bracket:

```text
)
]
}
```

we check the bracket currently at the top.

It must be the matching opening bracket.

### Example

```text
s = "{[]}"
```

Process:

```text
{
```

Push:

```text
{
```

Then:

```text
[
```

Push:

```text
{
[
```

Then:

```text
]
```

Top is:

```text
[
```

Match → Pop.

Then:

```text
}
```

Top is:

```text
{
```

Match → Pop.

The Stack is empty.

Therefore:

```text
True
```

## Output

## Brute Force

A brute-force approach would repeatedly search for adjacent matching pairs:

```text
()
[]
{}
```

and remove them.

For example:

```text
"{[]}"
```

First remove:

```text
[]
```

Then:

```text
{}
```

The string becomes empty.

However, repeatedly modifying and rescanning the string is inefficient.

The problem itself has a natural LIFO structure, so a Stack gives a direct solution.

## Optimal Approach

Use a Stack.

For every character:

### Opening bracket

If the character is:

```text
(
[
{
```

push it.

### Closing bracket

If the character is:

```text
)
]
}
```

then:

1. Check whether the Stack is empty.
2. Check whether the top is the matching opening bracket.
3. If it does not match, return `False`.
4. Otherwise pop the opening bracket.

After processing the entire string:

```text
Stack empty → True
Stack not empty → False
```

## Algorithm

```text
Create empty Stack

For every character in s:

    If character is an opening bracket:
        Push it

    Otherwise:

        If Stack is empty:
            return False

        If top does not match character:
            return False

        Pop the top

After the loop:

    If Stack is empty:
        return True

    Otherwise:
        return False
```

````python
class Solution:

    def isValid(self, s: str) -> bool:

        stack = []

        pairs = {
            ')': '(',
            ']': '[',
            '}': '{'
        }

        for char in s:

            if char in '([{':
                stack.append(char)

            else:

                if not stack:
                    return False

                if stack[-1] != pairs[char]:
                    return False

                stack.pop()

        return len(stack) == 0
````

## Complexity

Let `N` be the length of the string.

### Time

Every character is processed once.

```text
O(N)
```

### Space

In the worst case, all characters can be opening brackets.

```text
((([[[{{{
```

Therefore the Stack can contain `N` elements.

```text
O(N)
```

## Comparison

| Approach | Time | Space |
|---|---:|---:|
| Repeatedly remove matching pairs | O(N²) | O(N) |
| Stack | O(N) | O(N) |

The Stack approach is optimal in time because every character only needs to be processed once.

## Pattern Recognition

When a problem contains:

```text
Opening → Closing
```

and requires:

```text
Correct nesting
Correct order
Most recent opening must close first
```

think:

```text
STACK
```

Typical examples include:

```text
Parentheses
Brackets
Braces
Nested expressions
HTML/XML tags
```

The key signal is:

```text
Last opened
    ↓
First closed
```

That is LIFO.

## Takeaway

The important observation is not simply:

```text
"This is a bracket problem."
```

It is:

```text
The most recently opened bracket
must be closed first.
```

That is exactly Stack behavior.

```text
Opening bracket
      ↓
    PUSH
      ↓
    STACK
      ↓
    POP
      ↓
Matching closing bracket
```

This is one of the fundamental patterns for recognizing when a Stack should be used in an interview problem.
