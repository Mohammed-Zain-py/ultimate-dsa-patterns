# Stack Peek and Is Empty Operations

# Problem Statement

Implement the `peek()` and `is_empty()` operations of a Stack.

The Stack must support:

```text
peek()
is_empty()
```

`peek()` should return the element currently at the top without removing it.

`is_empty()` should determine whether the Stack contains any elements.

Example

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

`peek()`:

```text
30
```

The Stack remains:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

For `is_empty()`:

```text
TOP
 ↓
NULL
```

should return:

```text
True
```

# Problem Explanation

Unlike `pop()`, `peek()` does not modify the Stack.

It only looks at the top element.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Peek:

```text
30
```

After Peek:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Nothing was removed.

For a Linked List implementation, the top element is directly available through:

```python
self.top.value
```

The `is_empty()` operation checks whether the top pointer is `None`.

```text
TOP = None

↓

Stack is empty
```

# Brute Force

With a Python list, the top element can be accessed using:

```python
stack[-1]
```

and the Stack can be checked using:

```python
len(stack) == 0
```

Example:

```python
stack = [10, 20, 30]

print(stack[-1])

print(len(stack) == 0)
```

However, the Stack interface should expose these operations through methods such as:

```text
peek()

is_empty()
```

````python
stack = [10, 20, 30]

print("Top:", stack[-1])

print("Empty:", len(stack) == 0)
````

# Optimal Approach

For a Linked List Stack:

### Peek

Return:

```python
self.top.value
```

after checking that the Stack is not empty.

### Is Empty

Check:

```python
self.top is None
```

Both operations require direct access to the top.

No traversal is needed.

# Dry Run — Peek

Initial Stack:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Call:

```text
peek()
```

Access:

```text
TOP.value
```

Result:

```text
30
```

Stack after Peek:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

The Stack has not changed.

# Dry Run — Is Empty

### Case 1: Empty Stack

```text
TOP
 ↓
NULL
```

Check:

```text
TOP is None
```

Result:

```text
True
```

### Case 2: Non-Empty Stack

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Check:

```text
TOP is None
```

Result:

```text
False
```

````python
class Node:

    def __init__(self, value):

        self.value = value
        self.next = None


class Stack:

    def __init__(self):

        self.top = None

    def push(self, value):

        node = Node(value)

        node.next = self.top

        self.top = node

    def peek(self):

        if self.is_empty():

            return None

        return self.top.value

    def is_empty(self):

        return self.top is None
````

````python
stack = Stack()

print("Empty:", stack.is_empty())

stack.push(10)
stack.push(20)
stack.push(30)

print("Top:", stack.peek())

print("Empty:", stack.is_empty())
````

# Why Is Peek O(1)?

Consider:

```text
TOP
 ↓
30 → 20 → 10 → 5 → 2 → 1 → NULL
```

The top pointer already points directly to:

```text
30
```

Therefore, Peek only needs:

```python
self.top.value
```

It does not need to traverse:

```text
30 → 20 → 10 → 5 → 2 → 1
```

So:

```text
Peek → O(1)
```

The number of elements in the Stack does not affect the operation.

# Why Is Is Empty O(1)?

To determine whether the Stack is empty, we only check:

```python
self.top is None
```

There is no need to count or traverse the nodes.

Therefore:

```text
Is Empty → O(1)
```

# Edge Cases

### Empty Stack

```text
TOP
 ↓
NULL
```

`peek()` should not attempt:

```python
self.top.value
```

because `self.top` is `None`.

A safe implementation returns:

```text
None
```

### One Element

```text
TOP
 ↓
10 → NULL
```

`peek()` returns:

```text
10
```

The Stack remains unchanged.

### Multiple Elements

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

`peek()` returns:

```text
30
```

Only the top is inspected.

# Common Mistakes

### Mistake 1 — Using Pop for Peek

Incorrect:

```python
return self.pop()
```

This removes the top element.

Peek must not modify the Stack.

Correct:

```python
return self.top.value
```

---

### Mistake 2 — Traversing the Linked List

There is no reason to traverse:

```text
30 → 20 → 10
```

The top is already known.

---

### Mistake 3 — Forgetting the Empty Check

This can cause:

```text
AttributeError
```

when accessing:

```python
self.top.value
```

while:

```python
self.top is None
```

# Complexity

| Operation | Time | Extra Space |
|---|---:|---:|
| Peek | O(1) | O(1) |
| Is Empty | O(1) | O(1) |

Neither operation requires traversal.

Overall Stack storage remains:

```text
O(N)
```

# Comparison

| Operation | What It Does | Changes Stack? | Time |
|---|---|---|---:|
| Push | Adds top element | Yes | O(1) |
| Pop | Removes top element | Yes | O(1) |
| Peek | Views top element | No | O(1) |
| Is Empty | Checks whether Stack is empty | No | O(1) |

The important distinction is:

```text
Peek

↓

Look

Pop

↓

Remove
```

# Pattern Recognition

When a problem asks:

```text
What is the current top element?
```

Think:

```text
PEEK
```

When it asks:

```text
Does the Stack contain anything?
```

Think:

```text
IS EMPTY
```

Do not remove an element when the problem only asks you to inspect it.

# Takeaway

`peek()` and `is_empty()` are simple operations, but they are fundamental to almost every Stack implementation.

```text
Peek

↓

View TOP

↓

Do not remove
```

```text
Is Empty

↓

Check TOP == None
```

Both are:

```text
O(1)
```

The main rule to remember is:

```text
PEEK = LOOK

POP = REMOVE
```
