# Min Stack

# Problem Statement

Design a Stack that supports the following operations in constant time:

```text
push(value)
pop()
top()
getMin()
```

The special operation is:

```text
getMin()
```

It must return the minimum element currently present in the Stack.

All operations must run in:

```text
O(1)
```

### Example

Perform:

```text
push(-2)
push(0)
push(-3)
```

Stack:

```text
TOP
 ↓
-3
-0
-2
```

`getMin()` returns:

```text
-3
```

After:

```text
pop()
```

Stack:

```text
TOP
 ↓
0
-2
```

Now:

```text
getMin()
```

returns:

```text
-2
```

# Problem Explanation

A normal Stack already gives us:

```text
push() → O(1)
pop()  → O(1)
top()  → O(1)
```

The difficult operation is:

```text
getMin()
```

Suppose the Stack is:

```text
TOP
 ↓
3
5
2
7
1
```

The minimum is:

```text
1
```

But the minimum is not necessarily at the top.

If we simply scan the entire Stack every time:

```text
3
5
2
7
1
```

we need:

```text
O(N)
```

That violates the requirement.

So we need to maintain extra information while using the Stack.

The key idea is:

```text
Store the minimum value seen so far
```

This allows `getMin()` to directly return the current minimum.

# Input

The Stack receives a sequence of operations.

Example:

```text
push(-2)
push(0)
push(-3)
getMin()
pop()
getMin()
```

# Output

```text
-3
-2
```

The first `getMin()` returns:

```text
-3
```

After removing `-3`, the minimum becomes:

```text
-2
```

# Brute Force

The simplest approach is to use one normal Stack.

For:

```text
getMin()
```

traverse all elements and find the smallest value.

Example:

```text
TOP
 ↓
3
5
2
7
1
```

Scan:

```text
3 → minimum = 3
5 → minimum = 3
2 → minimum = 2
7 → minimum = 2
1 → minimum = 1
```

Return:

```text
1
```

The problem is that every `getMin()` requires:

```text
O(N)
```

This does not satisfy the requirement.

We need another approach.

# Optimal Approach

Use **two Stacks**.

```text
Main Stack
Min Stack
```

The Main Stack stores the actual values.

The Min Stack stores the minimum value corresponding to each level.

Example:

Push:

```text
5
```

Main:

```text
5
```

Min:

```text
5
```

Push:

```text
3
```

Main:

```text
3
5
```

Min:

```text
3
5
```

Push:

```text
7
```

Main:

```text
7
3
5
```

Min:

```text
3
3
5
```

The top of the Min Stack always represents:

```text
Current Minimum
```

Therefore:

```text
getMin()
```

simply returns:

```python
min_stack[-1]
```

which is:

```text
O(1)
```

# Why Does the Min Stack Work?

Suppose we push:

```text
5
3
7
2
```

After every Push:

```text
Main Stack       Min Stack

7                2
2                2
3                3
5                5
```

The top of the Min Stack is:

```text
2
```

Therefore:

```text
getMin() = 2
```

Now Pop `2`.

Both stacks must pop:

```text
Main Stack       Min Stack

7                2
3                3
5                5
```

The new minimum is:

```text
3
```

The Min Stack automatically restores the previous minimum.

That is the key idea.

# Algorithm

### Push

```text
Push value into Main Stack.

If Min Stack is empty:
    Push value into Min Stack.

Otherwise:
    Find the smaller of:
        value
        current minimum

    Push that minimum into Min Stack.
```

### Pop

```text
Pop from Main Stack.

Pop from Min Stack.
```

### Top

```text
Return top of Main Stack.
```

### Get Minimum

```text
Return top of Min Stack.
```

````python
class MinStack:

    def __init__(self):

        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:

        self.stack.append(val)

        if not self.min_stack:
            self.min_stack.append(val)

        else:
            current_min = min(val, self.min_stack[-1])

            self.min_stack.append(current_min)

    def pop(self) -> None:

        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:

        return self.stack[-1]

    def getMin(self) -> int:

        return self.min_stack[-1]
````

# Edge Cases

### Empty Stack

There is no minimum when the Stack contains no elements.

The implementation should only call:

```text
top()
getMin()
pop()
```

when the Stack is non-empty, as required by the problem constraints.

### One Element

```text
Main       Min

10         10
```

Therefore:

```text
getMin() = 10
```

### Duplicate Minimum

Consider:

```text
push(2)
push(2)
push(1)
push(1)
```

The Min Stack preserves the minimum at every level.

After popping one `1`, another `1` remains the minimum.

This is why storing the minimum at every level is useful.

# Common Mistake

A common approach is to maintain only one variable:

```python
self.minimum
```

This works for `getMin()` until the current minimum is popped.

Example:

```text
Stack:

5
3
2
```

Current minimum:

```text
2
```

Pop `2`.

Now:

```text
5
3
```

What is the new minimum?

```text
3
```

A single variable does not automatically remember that previous minimum.

The second Stack solves this problem by preserving the minimum at every level.

# Complexity

Let `N` be the number of elements in the Stack.

| Operation | Time | Extra Space |
|---|---:|---:|
| `push()` | O(1) | O(1) |
| `pop()` | O(1) | O(1) |
| `top()` | O(1) | O(1) |
| `getMin()` | O(1) | O(1) |

The Min Stack requires an additional Stack containing up to `N` values.

Therefore:

```text
Overall Space → O(N)
```

# Comparison

| Approach | `getMin()` | Overall Space |
|---|---:|---:|
| Scan normal Stack | O(N) | O(N) |
| Two Stacks | O(1) | O(N) |

The second approach uses more auxiliary state, but it satisfies the requirement that **every operation runs in O(1)**.

# Pattern Recognition

This problem is not just about finding a minimum.

The important requirement is:

```text
getMin() must be O(1)
```

Whenever a problem asks for an operation that would normally require repeatedly scanning a data structure, ask:

```text
Can I maintain additional information
while modifying the data structure?
```

Here:

```text
Normal Stack
    +
Minimum Stack
    ↓
O(1) minimum query
```

This is a general data-structure design pattern:

```text
Maintain extra state
to make future queries faster.
```

# Takeaway

A normal Stack cannot answer:

```text
"What is the minimum?"
```

in O(1) unless we maintain additional information.

The Min Stack solves this using:

```text
Main Stack
     +
Min Stack
```

The crucial invariant is:

```text
Top of Min Stack
        ↓
Current minimum of Main Stack
```

Therefore:

```text
push()   → O(1)
pop()    → O(1)
top()    → O(1)
getMin() → O(1)
```

The important interview lesson is:

> When a data structure must support a new query efficiently, consider maintaining auxiliary state rather than recomputing the answer every time.
