# How to Identify Stack Problems

# Why Pattern Recognition Matters

Knowing how a Stack works is only the beginning.

In an interview, the problem usually will not say:

```text
Use a Stack.
```

Instead, the problem describes a situation where

```text
The most recent item

must be handled first.
```

Your job is to recognize that behavior.

The goal is:

```text
Problem

↓

Observe the behavior

↓

Identify LIFO

↓

Think Stack
```

# The Core Question

When reading a problem, ask:

```text
Does the most recently encountered
or inserted element need to be processed first?
```

If the answer is yes:

```text
Think STACK
```

Example:

```text
Open:

(

Then:

(

Then:

[

```

When closing symbols appear, the most recently opened symbol must be handled first.

That is LIFO behavior.

# Signal 1 — Matching Parentheses

Problems involving:

```text
()

{}

[]
```

are strong Stack candidates.

Example:

```text
({[]})
```

The most recently opened bracket must be matched first.

```text
Open {

↓

Open [

↓

Close ]

↓

Close {
```

The last opening bracket is the first one resolved.

Therefore:

```text
LIFO

↓

STACK
```

# Signal 2 — Nested Structures

Nested structures naturally create LIFO behavior.

Example:

```text
A
└── B
    └── C
```

To finish processing:

```text
C

↓

B

↓

A
```

The deepest and most recently entered structure is handled first.

Common examples include:

```text
Nested brackets

Nested expressions

Function calls

Directory paths

HTML/XML tags
```

When a problem repeatedly enters and exits nested structures, think Stack.

# Signal 3 — Undo Operations

Consider:

```text
Action 1
Action 2
Action 3
```

If the user presses Undo:

```text
Action 3
```

must be removed first.

Then:

```text
Action 2
```

Then:

```text
Action 1
```

This is:

```text
Last Action

↓

First Undone
```

Therefore:

```text
LIFO

↓

STACK
```

# Signal 4 — Next Greater / Smaller Element

This is one of the most important Stack patterns.

Suppose:

```text
[2, 1, 5, 3, 4]
```

For each element, we may need to find:

```text
Next Greater Element
```

For `2`:

```text
2 → 5
```

For `1`:

```text
1 → 5
```

For `3`:

```text
3 → 4
```

The challenge is efficiently maintaining elements that are still waiting for their answer.

This leads to:

```text
Monotonic Stack
```

Important related problems:

```text
Next Greater Element

Next Smaller Element

Previous Greater Element

Previous Smaller Element

Daily Temperatures

Stock Span
```

# Signal 5 — Expression Problems

Stacks are commonly used when working with expressions.

Examples:

```text
Infix

A + B
```

```text
Postfix

A B +
```

```text
Prefix

+ A B
```

Common interview problems include:

```text
Evaluate Postfix

Evaluate Prefix

Infix to Postfix

Infix to Prefix

Basic Calculator
```

Why Stack?

Operators and operands often need to be processed according to a specific recent-first order.

# Signal 6 — Reverse Order Processing

Suppose we have:

```text
1 → 2 → 3 → 4
```

and need to process them in reverse:

```text
4 → 3 → 2 → 1
```

A Stack naturally provides this behavior.

```text
Push

1

2

3

4
```

Stack:

```text
TOP
 ↓
4
3
2
1
```

Pop:

```text
4 → 3 → 2 → 1
```

So whenever a problem requires:

```text
Reverse order

↓

Think Stack
```

However, do not automatically use a Stack if a simpler two-pointer or in-place approach exists.

# Signal 7 — Backtracking

Backtracking repeatedly makes a choice and later needs to undo it.

Conceptually:

```text
Choose

↓

Explore

↓

Undo

↓

Try Another Choice
```

The recent choice must be undone first.

That is Stack-like behavior.

Example:

```text
Choose A

↓

Choose B

↓

Choose C
```

Undo:

```text
C

↓

B

↓

A
```

Recursion performs this using the call stack.

An explicit Stack can also be used when implementing the process iteratively.

# Signal 8 — DFS

Depth First Search can be implemented using a Stack.

Suppose:

```text
      A
     / \
    B   C
   /
  D
```

DFS explores deeply before returning.

Conceptually:

```text
A

↓

B

↓

D

↓

Backtrack

↓

C
```

The nodes that still need to be explored are maintained using Stack behavior.

So:

```text
Depth First

↓

Think STACK
```

BFS is different:

```text
BFS → Queue

DFS → Stack
```

# Signal 9 — Remove Recent Items

Some problems repeatedly ask you to remove the most recent item.

Examples:

```text
Remove Last Added

Undo Last Action

Cancel Most Recent Operation

Remove Previous Element
```

The common behavior is:

```text
Add

Add

Add

↓

Remove Last
```

That is directly:

```text
LIFO
```

and therefore:

```text
STACK
```

# Monotonic Stack

A Monotonic Stack is not a different data structure.

It is a Stack maintained in a specific order.

It can be:

```text
Increasing

or

Decreasing
```

Example:

```text
Increasing Stack

1
3
5
7
```

When a new value violates the required order, elements are popped.

This is the key idea behind many difficult Stack problems.

Common problems:

```text
Next Greater Element

Next Smaller Element

Daily Temperatures

Stock Span

Largest Rectangle in Histogram
```

Recognizing these problems is one of the most important Stack skills.

# Stack Decision Tree

When reading a problem:

```text
Does the problem require
recent-first processing?
            |
           YES
            ↓
          STACK
            |
            ├── Matching?
            │      ↓
            │   Parentheses
            │
            ├── Nested?
            │      ↓
            │   Stack
            │
            ├── Next/Previous
            │   Greater/Smaller?
            │      ↓
            │   Monotonic Stack
            │
            ├── Expression?
            │      ↓
            │   Stack
            │
            ├── Undo/Backtrack?
            │      ↓
            │   Stack
            │
            └── DFS?
                   ↓
                 Stack
```

# Stack Recognition Cheat Sheet

| Problem Signal | Likely Pattern |
|---|---|
| Matching brackets | Stack |
| Nested structures | Stack |
| Undo / recent action | Stack |
| Reverse processing | Stack |
| Next Greater | Monotonic Stack |
| Next Smaller | Monotonic Stack |
| Previous Greater | Monotonic Stack |
| Previous Smaller | Monotonic Stack |
| Expression evaluation | Stack |
| DFS | Stack |
| Backtracking | Stack / Call Stack |
```

# Important Warning

Do not use a Stack simply because the problem contains the word:

```text
"previous"
```

or:

```text
"reverse"
```

First identify the actual requirement.

Ask:

```text
Why do I need LIFO?
```

If there is no clear LIFO behavior, another pattern may be better.

For example:

```text
Reverse Linked List

```

can be solved in:

```text
O(1) extra space
```

using pointer manipulation.

A Stack would work, but it would not be the optimal approach.
```

# Pattern Recognition Examples

### Example 1

```text
Check whether brackets are balanced.
```

Think:

```text
Opening bracket

↓

Remember it

↓

Closing bracket

↓

Match most recent opening
```

Pattern:

```text
STACK
```

---

### Example 2

```text
Find the next greater element for every element.
```

Think:

```text
Elements waiting for an answer
```

Pattern:

```text
MONOTONIC STACK
```

---

### Example 3

```text
Find the middle node of a Linked List.
```

Think:

```text
Fast + Slow
```

Not Stack.

Pattern:

```text
TWO POINTERS
```

# Final Recognition Rule

Do not memorize:

```text
Problem X → Stack
Problem Y → Stack
Problem Z → Stack
```

Instead memorize the behavior:

```text
Last In

↓

First Out
```

Then recognize its variations:

```text
Nested

Matching

Undo

Reverse

Recent-first

Next Greater / Smaller

Expression Processing

DFS
```

That is how you identify Stack problems in an interview.

# Takeaway

The most important Stack skill is not knowing:

```text
push()

pop()
```

It is recognizing when the problem requires:

```text
LIFO
```

The main interview patterns are:

```text
Basic Stack

↓

Parentheses

↓

Expression

↓

Monotonic Stack

↓

DFS / Backtracking
```

Once the LIFO behavior becomes obvious, choosing Stack becomes much easier.
