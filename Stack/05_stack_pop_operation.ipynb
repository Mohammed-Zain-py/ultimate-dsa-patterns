# Stack Pop Operation

## Problem Statement

Implement the `pop` operation of a Stack.

The operation must:

- Remove the element from the top of the Stack.
- Return the removed element.
- Update the top of the Stack.
- Handle an empty Stack safely.

Input

```text
Stack

TOP
 ↓
30 → 20 → 10 → NULL
```

Operation

```text
pop()
```

Output

```text
30
```

Remaining Stack

```text
TOP
 ↓
20 → 10 → NULL
```

Explanation

The element at the top is always removed first because a Stack follows:

```text
LIFO

Last In
First Out
```

## Problem Explanation

For a Linked List implementation, the top of the Stack is the head.

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

To remove `30`, we do not need to traverse the list.

Simply move:

```text
TOP

↓

TOP.next
```

Before:

```text
TOP
 ↓
30 → 20 → 10
```

After:

```text
TOP
 ↓
20 → 10
```

The removed node contains:

```text
30
```

So the operation has two responsibilities:

```text
1. Move TOP

2. Return removed value
```

## Brute Force

Using a Python list, `pop()` already performs the required operation.

```python
stack = [10, 20, 30]

value = stack.pop()

print("Popped:", value)

print("Stack:", stack)
```

The end of the list represents the top.

```text
[10, 20, 30]
         ↑
        TOP
```

````python
stack = [10, 20, 30]

value = stack.pop()

print("Popped:", value)

print("Stack:", stack)
````

## Complexity

For a Python list Stack:

```text
Time  → O(1)

Space → O(1)
```

For a Linked List Stack:

```text
Time  → O(1)

Space → O(1)
```

The top element can be removed directly without traversing the Stack.

## Optimal Approach

For a Linked List Stack:

```text
1. Check whether the Stack is empty.

2. Store the current top value.

3. Move TOP to TOP.next.

4. Decrease the size.

5. Return the stored value.
```

Visual:

```text
Before

TOP
 ↓
30 → 20 → 10 → NULL
```

Store:

```text
value = 30
```

Move TOP:

```text
TOP
 ↓
20 → 10 → NULL
```

Return:

```text
30
```

## Dry Run

Initial:

```text
TOP
 ↓
30 → 20 → 10 → NULL
```

Store:

```text
value = 30
```

Move:

```text
TOP = TOP.next
```

Now:

```text
TOP
 ↓
20 → 10 → NULL
```

Return:

```text
30
```

Second Pop:

```text
20
```

Stack:

```text
TOP
 ↓
10 → NULL
```

Third Pop:

```text
10
```

Stack:

```text
TOP
 ↓
NULL
```

````python
class Node:

    def __init__(self, value):

        self.value = value

        self.next = None


class Stack:

    def __init__(self):

        self.top = None

        self.size = 0

    def push(self, value):

        node = Node(value)

        node.next = self.top

        self.top = node

        self.size += 1

    def pop(self):

        if self.is_empty():

            return None

        value = self.top.value

        self.top = self.top.next

        self.size -= 1

        return value

    def is_empty(self):

        return self.top is None
````

````python
stack = Stack()

stack.push(10)
stack.push(20)
stack.push(30)

print("Popped:", stack.pop())

print("Popped:", stack.pop())

print("Popped:", stack.pop())
````

## Edge Cases

### Empty Stack

```text
TOP
 ↓
NULL
```

Calling:

```text
pop()
```

should not attempt:

```text
TOP.value
```

because `TOP` is `None`.

Return:

```text
None
```

### One Element

```text
TOP
 ↓
10 → NULL
```

Pop:

```text
10
```

After:

```text
TOP
 ↓
NULL
```

The Stack becomes empty.

## Common Mistake

Incorrect:

```python
self.top = self.top.next

return self.top.value
```

This returns the **new** top instead of the removed value.

Example:

```text
30 → 20 → 10
```

After moving TOP:

```text
20 → 10
```

Returning `self.top.value` gives:

```text
20
```

But the removed value was:

```text
30
```

Correct approach:

```python
value = self.top.value

self.top = self.top.next

return value
```

## Takeaway

Pop means:

```text
Remove TOP
```

For a Linked List Stack:

```text
Save Value

↓

Move TOP

↓

Return Value
```

The operation is:

```text
O(1)
```

The most important detail is to save the old top value **before** changing the top pointer.
