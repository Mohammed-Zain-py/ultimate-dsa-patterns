# Monotonic Stack Cheat Sheet

Next Greater:

pop while:

stack[-1] <= current


Next Smaller:

pop while:

stack[-1] >= current


Previous Greater:

pop while:

stack[-1] <= current


Previous Smaller:

pop while:

stack[-1] >= current

# Direction Cheat Sheet

Previous:

answer is on the left

↓

process:

left → right


Next:

answer is on the right

↓

process:

right → left

# Greater / Smaller Table

```text
Problem              Direction       Pop Condition

Next Greater         Right → Left    <= current

Next Smaller         Right → Left    >= current

Previous Greater     Left → Right    <= current

Previous Smaller     Left → Right    >= current

Remember:
The pop condition removes
elements that cannot be
the required answer.
```

# Common Stack Mistakes

Mistake 1:

Using the wrong direction.


Mistake 2:

Using `>` when the problem
requires `>=`.


Mistake 3:

Forgetting that equal values
may need to be removed.


Mistake 4:

Reversing operand order in
expression problems.


Mistake 5:

Forgetting to process
remaining Stack elements.

# Interview Explanation Template

When explaining a Stack solution:

1. Explain why the Stack
   matches the problem behavior.

2. Explain what the Stack
   stores.

3. Explain when elements
   are pushed.

4. Explain when elements
   are popped.

5. Explain what the Stack top
   represents.

6. Give time and space
   complexity.
